```python
# Respuesta

# a)
lista_incorrecta = [23, False, 6.6, "nohtyp", "cou", True, 1345, "aloh"]

# Primer error: no se había definido lista_nueva antes de intentar añadir
# elementos a ella.
lista_nueva = []

# Correciones:
# El [:-1:] del código original es incorrecto para invertir un string.
# En su lugar, [::-1] invierte correctamente el orden.
# .upper() es lo que convierte el string a mayúsculas.

# En el bucle, si el elemento es de tipo string,
# lo invertimos y pasamos a mayúsculas.
for element in lista_incorrecta:
    if isinstance(element, str):
        lista_nueva.append(element[::-1].upper())

print("a)")
print(lista_nueva)


# b) Ponemos el código en forma de list comprehension. Usamos if-else
# para elegir entre invertir y transformar el string o bien dividir el número.
# Usamos isinstance(element, str) para detectar strings.
print("\nb)")
lista_nueva = [element[::-1].upper() for element in lista_incorrecta if isinstance(element, str)]
print(lista_nueva)


# c.1) La lista para almacenar strings transformados y números divididos entre
# 2. Para ello, identificamos todos los enteros y decimales y simplemente
# dividimos cada uno por 2.
lista_nueva = []

for element in lista_incorrecta:
    if isinstance(element, str):
        lista_nueva.append(element[::-1].upper())
    elif isinstance(element, (int, float)):
        lista_nueva.append(element / 2)

print("\nc.1)")
print(lista_nueva)

# c.2) Ahora creamos una list comprehension con strings transformados y
# números divididos por 2. Usamos isinstance(element, (int, float)) para
# detectar si el elemento es numérico y luego dividorlo por 2.
# Usamos isinstance(element, str) para detectar strings y revertirlos.
lista_nueva = [element[::-1].upper() if isinstance(element, str) else element / 2
    for element in lista_incorrecta if isinstance(element, (str, int, float))]
print("\nc.2)")
print(lista_nueva)
```

```python
# Respuesta

# a)
lista1 = ["Pedro", 27, "Alex", False, "ari", 5, True, "john", 9.9, "SILVIA", False, True, "Xavi", "marIA", 104, "Tim"]

# Usamos lista_tipos para almacenar el primer elemento de cada tipo
lista_tipos1 = []

# Usamos un bucle para recorrer la lista y agregar el primer elemento de cada
# tipo. Usamos un set (i.e., conjunto) tipos_encont para apuntar los tipos que
# ya hemos agregado a lista_tipos, y los que no estén, se irán agregando
# en el bucle. A diferencia de listas o tuplas, los sets no tienen un orden
# específico y tampoco permiten elementos repetidos, lo cuál nos viene perfecto.
tipos_encont1 = set()
for element in lista1:
    if type(element) not in tipos_encont1:
        lista_tipos1.append(element)
        tipos_encont1.add(type(element))

print("a)")
print(lista_tipos1)

 # b) La función recibe una lista de elementos mixtos y devuelve una lista con
 # el 1r elemento de cada tipo usando el bucle for y set para apuntar los
 # tipos ya añadidos a la lista de resultados. La ventaja clave aquí es que
 # la función acepta cualquier lista fácilmente con un simple cambio de nombre
 # del argumento (poniendo directamente el nombre de la lista).

def primeros_elem_por_tipo(lista):
    lista_tipos2 = []
    tipos_encont2 = set()

    for element in lista:
        if type(element) not in tipos_encont2:
            lista_tipos2.append(element)
            tipos_encont2.add(type(element))

    return lista_tipos2

# Llamada a la función
print("\nb)")
print(primeros_elem_por_tipo(lista1))

# c) El primer element indica el elemento actual de lista1 que vamos a añadir
# a la nueva lista. El element dentro del for es el iterador que recorre lista.
# Verificamos si el tipo del elemento aún no ha sido agregado al conjunto
# mediante "if type(element) not in tipos_encont3".
# El "and not tipos_encontrados.add(type(element))" sirve
# para agregar el elemento solo si su tipo no está en
# tipos_encontrados. La función .add() devuelve None (i.e., un False),
# permitiendo que la evaluación siga.

# Así, la primera parte (not in) verifica si el tipo no está en el
# conjunto, dando True o False.
# La segunda parte (and not .add()) se usa solo si la primera es True,
# y añade el tipo al conjunto y siempre devuelve True (not None/False).
# Así, es True global solo si ambas partes son True. Si el tipo ya está en el
# conjunto, la primera parte falla y el elemento no se añade.


tipos_encont3 = set()
lista_tipos3 = [element for element in lista1 if type(element) not in
               tipos_encont3 and not tipos_encont3.add(type(element))]

print("\nc)")
print(lista_tipos3)


# d)

# El código del apartado a) es más fácil de entender y mantener porque se usa
# un for explícito y la lógica es más clara. La versión del apartado c)
# es más compacta, pero puede ser más difícil de interpretar por la lógica
# compleja dentro de la list comprehension. Así, prefiero la versión a).
```

## Ejercicio 4

En este ejercicio vamos a escribir un código para calcular, de manera simplificada, las retenciones de IRPF de la nómina a partir de un salario anual bruto y su correspondiente salario anual neto.

La retención total del IRPF no se calcula aplicando un único porcentaje al salario bruto. En su lugar, se divide en tramos, y cada tramo de ingresos tributa a su tipo correspondiente. Esta progresividad se explica en la tabla siguiente:

| **Tramo de renta (€)**        | **Tipo impositivo (%)**|
|-------------------------------|------------------------|
| Hasta 12.450€                 | 19%                    |
| De 12.451€ a 20.200€          | 24%                    |
| De 20.201€ a 35.200€          | 30%                    |
| De 35.201€ a 60.000€          | 37%                    |
| De 60.000€ a 300.000€          | 45%                    |
| A partir de 300.000€           | 47%                    |

Si consideramos un salario de 40.000€, el cálculo de la retención de IRPF sería el siguiente:

- Los primeros 12.450€ de los ingresos anuales tributan al 19%.
- Los siguientes 7.750€ (de 12.451€ a 20.200€) tributan al 24%.
- Los siguientes 15.000€ (de 20.201€ a 35.200€) tributan al 30%.
- El resto, hasta el salario total de 40.000€, tributa al 37%.

A la hora de calcular estas retenciones, se tienen en cuenta varios factores a parte del salario, como discapacidades, situación familiar, duración del contrato, etc. Sin embargo, para este ejercicio, vamos a usar solo la información de la tabla para simplificar los cálculos.

**(3 puntos)**

a) Escribe un código que calcule el **salario neto anual** y **el porcentaje de la retención de IRPF total (%)** (es decir, el retenciones anuales / salario bruto anual). También se tiene que mostrar por pantalla un mensaje con la siguiente información:

`Un sueldo bruto de X tiene una retención de IRPF total del Y%. Por lo tanto, el sueldo neto es Z`

**(1.5 puntos)** $\color{green}{\text{NM}}$

b) Pon este código en forma de función con las siguientes características:
- Se tiene que pedir al usuario por pantalla que introduzca el sueldo bruto anual que quiere analizar.
- La función tiene que devolver como output el salario neto anual y el porcentaje de la retención de IRFP total.
- Si el salario bruto anual es menor a 15.000€, se avisa al usuario que no se le aplicará ninguna retención de IRPF, de modo que se devuelve el salario bruto como resultado, junto con un IRFP de 0.
- Con tus conocimientos de gestión de errores, tienes que controlar los siguientes problemas:
    - Si se especifica un salario anual bruto negativo, se tiene que mostrar un mensaje de error para avisar al usuario.
    - Si se introduce un valor por pantalla, se tiene que gestionar y mostrar un mensaje de error.

**(1.25 puntos)** $\color{green}{\text{NM}}$

c) Para comprobar que el programa funciona correctamente, muestra un ejemplo de cada situación:

- Se introduce un salario de 40.000€.
- Se introduce un salario inferior a 15.000€.
- Se introduce un valor erróneo como input (e.g. "hola").

**(0.25 puntos)** $\color{green}{\text{NM}}$

```python
# Respuesta

# a) El tipo impositivo es el % de impuestos que se aplica por tramo (19%...).
# La retención es la cantidad de dinero que efectivamente se paga como impuesto,
# calculado según los tramos y sus tipos impositivos.

# Definimos el salario bruto, los tramos de renta y los tipos impositivos
salario_bruto = 10000
tramos = [(12450, 0.19), (20200, 0.24), (35200, 0.30), (60000, 0.37), (300000, 0.45), (float('inf'), 0.47)]

# Creamos variables para la retención total y el salario restante que se grava
retenc_total = 0
salario_rest = salario_bruto

# Ahora calculamos la retención del IRPF aplicando los tipos de cada tramo
for x in range(len(tramos)):
    # Determinamos el límite inferior del tramo actual, donde para el primer
    # tramo el límite inferior es 0 y para los otros, el límite inferior es el
    # límite superior del tramo anterior.
    # "tramos[x - 1]" selecciona la tupla correspondiente al tramo anterior y
    # el [0] accede al límite superior de ese tramo anterior
    # i.e., el primer valor de la tupla (en nuestro caso el límite superior del
    # tramo de renta), lo cual se convierte en el límite inferior del tramo
    # actual. Así, básicamente accedemos al primer valor ([0]) de la
    # tupla anterior (tramos[x - 1]).
    if x == 0:
        limite_inf = 0
    else:
        limite_inf = tramos[x - 1][0]

    # Obtenemos el límite superior y el porcentaje impositivo del tramo actual
    limite_sup = tramos[x][0]  # I.e., el 1r valor de la tupla (límite superior)
    porcentaje = tramos[x][1]  # I.e., el 2n valor de la tupla (porcentaje)


    # Aquí verificamos si el salario bruto excede el límite inferior del tramo.
    # Si da el caso, calculamos la cantidad que se tiene que gravar en el tramo,
    # el cual es el valor más menor entre el salario restante y el rango del
    # tramo (limite_sup - limite_inf).

    if salario_bruto > limite_inf:
        cant_tributable = min(salario_rest, limite_sup - limite_inf)

        # Ahora calculamos la retención para este tramo y la sumamos a la
        # retención total
        retenc_total += cant_tributable * porcentaje

        # Ahora quitamos del salario restante lo ya gravado en este tramo
        salario_rest -= cant_tributable

    # Cerramos el bucle si no queda salario por gravar
    if salario_rest <= 0:
        break

# Obtenemos el salario neto restando la retención total del salario bruto
salario_neto = salario_bruto - retenc_total

# Obtenemos el % de retención total en relación al salario bruto
porcentaje_retenc = (retenc_total / salario_bruto) * 100

# Imprimimos los resultados
print("a)")
print(f"Un sueldo bruto de {salario_bruto}€ tiene una retención de IRPF total del {porcentaje_retenc:.2f}%. Por lo tanto, el sueldo neto es {salario_neto:.2f}€")
```

```python
# b) Definimos la función calcular_irpf(). Usamos try-except para manejar
# posibles errores que surgan al ejecutar ciertas partes del código. Así,
# el bloque try ejecuta el código que contiene. Si todo va bien, el programa
# continúa con normalidad. Si ocurre un error dentro del bloque try,
# Python maneja ese error con el bloque except.

def calcular_irpf():
    try:
        # Solicitamos introducir al usuario el salario bruto anual
        salario_bruto = float(input("Introduce el salario bruto anual (€): "))

        # Verificamos si el salario es negativo. Si da el caso, mostramos error.
        if salario_bruto < 0:
            print("Error: el salario introducido es negativo.")
            return

        # Si el salario bruto es menor a 15,000€, no aplica la retención de IRPF
        if salario_bruto < 15000:
            print("No se aplica retención de IRPF para salarios menores a 15.000€.")
            print(f"El salario neto es {salario_bruto}€ y la retención de IRPF total es 0%.")
            return

        # Ahora definimos los tramos de renta (límite superior de cada tramo)
        # y los tipos impositivos correspondientes
        tramos = [(12450, 0.19), (20200, 0.24), (35200, 0.30), (60000, 0.37),
            (300000, 0.45), (float('inf'), 0.47)]

        # Creamos variables para la retención total y el salario restante que
        # se grava
        retenc_total = 0
        salario_rest = salario_bruto

        # Ahora calculamos la retención del IRPF aplicando los tipos de
        # cada tramo
        for x in range(len(tramos)):
        # Determinamos el límite inferior del tramo actual, donde para el primer
        # tramo el límite inferior es 0 y para los otros, el límite inferior es
        # el límite superior del tramo anterior.
        # "tramos[x - 1]" selecciona la tupla correspondiente al tramo anterior
        # y el [0] accede al límite superior de ese tramo anterior
        # i.e., el primer valor de la tupla (en nuestro caso el límite superior
        # del tramo de renta), lo cual se convierte en el límite inferior del
        # tramo actual. Así, básicamente accedemos al primer valor ([0]) de la
        # tupla anterior (tramos[x - 1]).
            if x == 0:
                limite_inf = 0
            else:
                limite_inf = tramos[x - 1][0]

    # Obtenemos el límite superior y el porcentaje impositivo del tramo actual
            limite_sup = tramos[x][0]  #I.e., el 1r valor de la tupla (lím. sup.)
            porcentaje = tramos[x][1]  #I.e., el 2n valor de la tupla (%)

            # Aquí verificamos si el salario bruto excede el límite inferior del
            # tramo. Si da el caso, calculamos la cantidad que se tiene que
            # gravar en el tramo, el cual es el valor más menor entre el salario
            # restante y el rango del tramo (limite_sup - limite_inf).
            if salario_bruto > limite_inf:
                cant_tributable = min(salario_rest, limite_sup - limite_inf)

                # Ahora calculamos la retención para este tramo y la
                # sumamos a la retención total
                retenc_total += cant_tributable * porcentaje

                # Ahora quitamos del salario restante lo ya
                # gravado en este tramo
                salario_rest -= cant_tributable

            # Cerramos el bucle si no queda salario por gravar
            if salario_rest <= 0:
                break

        # Obtenemos el salario neto restando la retención total del salario bruto
        salario_neto = salario_bruto - retenc_total

        # Obtenemos el % de retención total en relación al salario bruto
        porcentaje_retenc = (retenc_total / salario_bruto) * 100

        # Imprimimos el resultado
        print(f"Un sueldo bruto de {salario_bruto:.2f}€ tiene una retención de IRPF total del {porcentaje_retenc:.2f}%. Por lo tanto, el sueldo neto es {salario_neto:.2f}€")

    except ValueError:
        # Mostramos un mensaje de error si el usuario introduce un
        # valor no numérico
        print("Error: el valor introducido no es válido. Introduzca un número.")

# Aquí llamamos a la función, es donde empieza todo.
calcular_irpf()
```
