## Ejercicio 2

2.1 Cread un diccionario usando Dictionary Comprehensions tal que cada relación clave-valor sea cada una de las descripciones cortas y el valor el número de registros en que aparece esta descripción corta:

   * Las claves correspondan a la descripciones cortas en español de la sanción (**Infraccio_Desc_Curta_ES**).  
   * Sus valores correspondan al número de registros de cada descripción corta

2.2 Mostrad las 15 primeras entradas del diccionario (clave y valor) una vez lo hemos ordenado según el nombre de registros que contienen cada número (de mayor a menor).

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1 punto)**

<div style="background-color: #fcf2f2; border-color: #dfb5b4; border-left: 5px solid #dfb5b4; padding: 0.5em;">
<strong>Solución</strong>
</div>

```python
# 2.1 Primero usamos value_counts() para contar las ocurrencias en
# "Infraccio_Desc_Curta_ES". Así, obtenemos claves que son descripcions y
# valores (frecuencia), que representan la cantidad de repeticiones por
# cada clave. Convertimos el resultado a diccionario
descripcion_corta_dict = data['Infraccio_Desc_Curta_ES'].value_counts().to_dict()

# 2.2 Ordenamos el diccionario por los valores (i.e., frecuencia, que es el
# segundo elemento de cada tupla mediante "key=lambda item: item[1]"),
# y seleccionamos las primeras 15 entradas. Luego, imprimimos las primeras 15
# descripciones más comunes y su número de ocurrencias

primeras_15 = sorted(descripcion_corta_dict.items(), key=lambda item: item[1],
                     reverse=True)[:15]

# Procedemos a mostrar las 15 primeras entradas ordenadas del diccionario.
print("Las 15 descripciones más frecuentes:")
for descripcion, ocurrencias in primeras_15:
    print(f"{descripcion}: {ocurrencias}")
```

```python
# Cargamos de nuevo el dataframe original del ejercicio 1 (no modicado)
data = pd.read_csv('/content/drive/My Drive/Colab_Notebooks/Programación_DS/Activity_1/data/denuncies_sancions.csv', encoding='latin1', sep=';')

# Definimos la función verificar_infraccion, donde data el parametro que
# representa el dataframe, codigo_infraccion es el código concreto introducido,
# y la normativa, que tiene el valor predeterminado LSV.


def verificar_infraccion(data, codigo_infraccion, normativa='LSV'):
    # Filtramos el dataframe para obtener la fila correspondiente al
    # código de infracción. Convertimos la columna Codi y codigo_infraccion
    # a string usando astype(str) y str() para evitar errores en el filtro
    # cuando los tipos de datos no coinciden. Así, aseguramos que ambos valores
    # se comparen correctamente independientemente de sus tipos originales.

    filtro = data['Infraccio_Codi'].astype(str) == str(codigo_infraccion)
    infraccion = data[filtro]

    # Verificamos si la infracción existe en el dataframe (para evitar errores)
    if infraccion.empty:
        print(f"No se encontró ninguna infracción con el código {codigo_infraccion}.")
        return False

    # Extraemos el valor de las columna Normativa y Data_inici_vigencia
    # de la fila filtrada, al haber una única coincidencia.
    normativa_infraccion = infraccion['Normativa'].values[0]
    fecha_vigencia = infraccion['Data_Inici_Vigencia'].values[0]

    # Comparamos el valor de normativa_infraccion obtenido del dataframe con el
    # parámetro normativa proporcionado a la función.
    # Si la condición es correcta, se devuelve el mensaje del if.
    if normativa_infraccion == normativa:
        print(f"Para la infracción con código '{codigo_infraccion}' con fecha de vigencia '{fecha_vigencia}' cumple la normativa '{normativa}'.")
    else:
        print(f"Para la infracción con código '{codigo_infraccion}' con fecha de vigencia '{fecha_vigencia}' no cumple la normativa '{normativa}'.")


# Llamamos a la función con un código de infracción específico y imprimimos.
# Omitir el parámetro de normativa para usar el valor por defecto
codigo_infraccion = '910'
verificar_infraccion(data, codigo_infraccion)

```

5.3 Utilizando una dict comprehension crea un diccionario donde las claves sean los códigos de infracción (Infraccio_Codi) y los valores sean las descripciones largas de la infracción en catalán (Infraccio_Desc_Llarga_CA), pero solo para aquellas infracciones en las que se hayan retirado puntos al conductor (Punts_A_Retirar > 0)

```python
# 5.1 Encontramos la infracción con el importe nominal más alto
# mediante .idxmax. Luego, vamos a su fila mediante .loc y la mostramos.
idx_importe_max = data['Import_Nominal_?'].idxmax()
infraccion_importe_max = data.loc[idx_importe_max]
infraccion_importe_max
```

```python
# 5.2 Definimos una función que busque infracciones por palabra clave en
# "Infraccio_Desc_Curta_ES". Para ello, es necesario poner de parametro un
# presunto dataframe.
def filtrar_infraccion(df_infr_filt: pd.DataFrame, palabra: str) -> pd.DataFrame:
    # Guardamos toda la columna de "Infraccio_Desc_Curta_ES" de "descripcion".
    # Luego, filtramos las infracciones que contienen "palabra" en "descripción"
    # case=False indica que la búsqueda no distingue entre casings.
    # na=False ignora los valores nulos.
    # regex=False sirve para tratar la palabra como texto literal
    descripcion = df_infr_filt['Infraccio_Desc_Curta_ES']
    resultados = df_infr_filt[descripcion.str.contains(palabra, case=False, na=False, regex=False)]

    # Mostramos las coincidencias
    if resultados.empty:
        print(f"No se encontraron infracciones que contengan la palabra '{palabra}'.")
    else:
        return resultados

# Llamamos a la función
filtrar_infraccion(data, "Libertad")
```

```python
# 5.3 Creamos un diccionario donde las claves son los códigos de infracción
# y los valores son las descripciones largas. Luego iteramos sobre cada fila
# del dataframe y guardamos aquellas filas en las que "Punts_A_Retirar" > 0
# puntos a retirar mayores que cero.
infracciones_dict = {row['Infraccio_Codi']: row['Infraccio_Desc_Llarga_CA']
    for index, row in data.iterrows() if row['Punts_A_Retirar'] > 0}

# Mostramos el diccionario completo mediante el bucle for, que itera sobre los
# elementos del diccionario.
print("Diccionario de infracciones con puntos a retirar:")
for codi, descripcion in infracciones_dict.items():
    print(f"Código: {codi}; Descripción: {descripcion}")
```

```python
# 6.1 Utilizamos str.contains() para buscar la palabra "velocitat" en la columna
# "Infraccio_Desc_Llarga_CA" sin importar el casing y ignorando los nulos.
# La expresión regular es "velocitat", mientras que la secuencia de escape \w*
# (que en este caso coincide con cero o más caracteres alfanuméricos)
# admite cualquier forma de la palabra (aunque no parece ser necesario para este
# caso). Además, el uso de la r asegura que las secuencias de escape se lean
# como son de forma nativa en vez de como texto literal.
descrip_larga = data['Infraccio_Desc_Llarga_CA']
coincidencias_velocitat = descrip_larga.str.contains(r'velocitat\w*', case=False, na=False)
infracciones_velocitat = descrip_larga[coincidencias_velocitat]
print("Infracciones que mencionan 'velocitat':")
print(infracciones_velocitat)
```

```python
# 6.2 Para encontrar infracciones que contienen números, buscamos cualquier
# número en la columna "Infraccio_Desc_Curta_ES".
# La expresión regular "\d+" busca una o más cifras. Usamos otra vez na=False
# para manejar los valores nulos para evitar errores.
descrip_corta = data['Infraccio_Desc_Curta_ES']
coincidencias_numeros = descrip_corta.str.contains(r'\d+', na=False)
infracciones_numeros = descrip_corta[coincidencias_numeros]
print("Infracciones que contienen números:")
print(infracciones_numeros)
```

## Ejercicio 7. Proyecto
**Ejercicio: Gestión de una lista de infracciones pendientes**

**Enunciado**:

Crea un programa que mantenga una lista de infracciones pendientes de tramitar. Solamente contendrá códigos de infracción (1 columna)

*   Comienza con una lista vacía.
*   Pide al usuario que añada infracciones (códigos de infracción) a la lista.
*   Permite al usuario quitar una infracción de la lista si ya ha sido tramitada.
*   Muestra la lista de infracciones pendientes después de cada operación.


**Detalles**:

*   El programa ofrecerá (pedirá al usuario) 3 opciones: Añadir infracción, eliminiar infracción y salir.
*   Si elige la opción Añadir, el programa debe permitir añadir múltiples infracciones hasta que el usuario decida terminar.
*   Si elige la opción Eliminiar, el usuario puede eliminar una infracción especificando su código. Si el código de infracción a eliminar no está en la lista, el programa debe notificar al usuario.
*   Si elige la opción Salir, el programa finaliza.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 puntos)**

<div style="background-color: #fcf2f2; border-color: #dfb5b4; border-left: 5px solid #dfb5b4; padding: 0.5em;">
<strong>Solución</strong>
</div>

```python
def gestionar_infrac():
    # Empezamos con crear una lista vacía para manejar las infracciones.
    infrac_pendientes = []

    # Ahora creamos un bucle infinito para manejar las opciones resaltadas.
    while True:
        print("1. Añadir infracción")
        print("2. Eliminar infracción")
        print("3. Salir")
        opcion = input("Elige una opción (1, 2, 3) --> ")

        # Si el usuario introduce 1, continuamos por aquí
        if opcion == "1":

            # Añadimos infracción mediante .append.
            # Si se escribe salir, se termina el programa.
            while True:
                codigo = input("Introduce código de infracción. Escribe salir para terminar --> ")
                if codigo == 'salir':
                    break
                infrac_pendientes.append(codigo)

        # Si el usuario introduce 2, continuamos por aquí
        elif opcion == "2":

            # Eliminamos la infracción ya introducida por el usuario
            # mediante .remove
            codigo = input("Código de infracción a eliminar --> ")
            if codigo in infrac_pendientes:
                infrac_pendientes.remove(codigo)
            else:
                print("No está en la lista.")

        # Si el usuario introduce 3, continuamos por aquí.
        # Imprimimos "Saliendo" y terminamos el programa.
        elif opcion == "3":
            print("Saliendo.")
            break

        # Mostramos la lista de infracciones tras cada operación
        print("Infracciones pendientes --> ", infrac_pendientes)

# Llamamos a la función para empezar a ejecutar el programa desde aquí.
gestionar_infrac()
```
