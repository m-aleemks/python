### Ejercicio 3

Crear un programa que nos indique la temperatura del día a partir de la temperatura mínima y máxima.

- Se preguntaran dos valores al usuario(temperatura mínima y máxima)
- Mostrará por pantalla ambos valores indicando cual es la mínima y cual es la máxima
- Calcular la media de ambos valores
- El mensaje final del programa deberá indicar si la temperatura media es baja (10 o menos grados), media (desde 10 a 20 grados) o alta (mayor de 20 grados)

**(1.5 puntos)** $\color{green}{\text{NM}}$

````python
# Respuesta

# Pedimos introducir la temperatura mínima. Convertimos ese input
# (número) de string a entero
temperatura_minima = int(input("Introduce la temperatura mínima: "))

# Solicitamos introducir la temperatura máxima
# La convertimos a int() para consistencia
temperatura_maxima = int(input("Introduce la temperatura máxima: "))

# Mostramos ambos valores y cuál es la mínima y la máxima
print("La temperatura mínima es:", temperatura_minima)
print("La temperatura máxima es:", temperatura_maxima)

# Procedemos a calcular la media de las temperaturas
media_temperatura = (temperatura_minima + temperatura_maxima) / 2

# Mostramos la temperatura media en la pantalla
print("La temperatura media es:", media_temperatura)

# Determinamos si la temperatura media es baja, media o alta
# Para ello, usamos las condiciones "if; elif; else" y las comparaciones "<; =;
# >" para evaluar el rango de la temperatura media

# 10 o menos grados
if media_temperatura <= 10:
    print("La temperatura media es baja.")

# Más de 10 y hasta 20 grados (elif = else if, para condiciones)
elif 10 < media_temperatura <= 20:
    print("La temperatura media es media.")

# Más de 20 grados (else espara el rango restante)
else:
    print("La temperatura media es alta.")
````

### Ejercicio 4

Considera el siguiente código:
```
notas = [6,8,2,5,4]
```
Estas son las notas de un estudiante y corresponden a las asignaturas Matemáticas, Física, Economía, Historia y Arte.

1. ¿Cúal es la nota media del alumno?
2. El profesor de Economía ha rectificado su nota y ahora es de un 6. ¿Cúal es la nueva nota media?
3. Con la nota actualizada, elimine las notas de las materias suspendidas (inferiores a 5).
4. Crea un diccionario con las asignaturas y las notas e indica la asignatura con la nota más baja y la asignatura con la nota más alta.

**NOTA :** Este ejercicio se ha de realizar modificando la lista original y no creando una nueva

**(1.5 puntos)** $\color{orange}{\text{EG}}$

````python
# Respuesta

# Definimos la lista de notas
# Corresponde a Matemáticas, Física, Economía, Historia, y Arte respectivamente
notas = [6, 8, 2, 5, 4]

# 1. Calculamos la nota media sumando las notas mediante función sum()
# y dividiendo por el número de notas obtenidas mediante la función len()
nota_media = sum(notas) / len(notas)
print("La nota media es:", nota_media)

# 2. Economía corresponde al índice 2. Así, actualizamos esa posición a 6
# y recalculamos la media
notas[2] = 6
nota_media_actualizada = sum(notas) / len(notas)
print("La nueva nota media del alumno es:", nota_media_actualizada)

# 3. Filtramos las notas suspendidas (menores a 5) mediante la función filter.
# La lambda es una función anónima, y aquí se usa para recorrer con condicional
notas = list(filter(lambda x: x >= 5, notas))
print("Notas después de eliminar materias suspendidas:", notas)

# Para crear un diccionario con las asignaturas y las notas
# primero definimos las asignaturas correspondientes. Conservamos Arte en la
# lista aunque quede suspendida, ya que no afectará el resultado
asignaturas = ["Matemáticas", "Física", "Economía", "Historia", "Arte"]

# Luego creamos el diccionario combinando las asignaturas y las notas mediante
# la función dict, y juntamos cada iteración de asignaturas y notas
# respectivamente. Al final, procedemos a imprimir el resultado
diccionario_notas = dict(zip(asignaturas, notas))
print("Diccionario de asignaturas y notas:", diccionario_notas)

# Ahora determinamos la asignatura con la nota más baja y la nota más alta
# mediante las funciones max y min, y procedemos a imprimir lo obtenido
asignatura_max_nota = max(diccionario_notas, key=diccionario_notas.get)
print("La asignatura con la nota más alta es:", asignatura_max_nota)
asignatura_min_nota = min(diccionario_notas, key=diccionario_notas.get)
print("La asignatura con la nota más baja es:", asignatura_min_nota)
````

### Ejercicio 5

Considera este listado de asignaturas:

| Asignaturas  | Crédito  |
|--------------|----------|
| Programación |6         |
| Estadística  |3         |
| Cálculo      |4         |
| Álgebra      |5         |

Realiza:

1) ¿Qué tipo de variable es la más adecuada para introducir la información? **(0.25 puntos)**

2) Muestra por pantalla todos los valores, por un lado, y todas las claves por otro **(0.25 puntos)**

3) Cálcula el total de créditos del curso. **(0.5 puntos)**

4) Elimina la entrada de `Álgebra`, y añade a `Trigonometria`con un valor de 2, muestra el resultado final **(0.5 puntos)**

**NOTA:** Es posible que tengas que consultar las funciones que proporcionan los diccionarios de Python en la [documentación oficial](https://docs.python.org/3/library/stdtypes.html#mapping-types-dict).

 **(1.5 puntos)** $\color{green}{\text{NM}}$

````python
# Respuesta
````


1) Un diccionario, ya que las claves pueden ser las asignaturas y los valores los créditos.

````python
# 2) Primero creamos el diccionario con asignaturas y créditos
asignaturas = {"Programación": 6, "Estadística": 3, "Cálculo": 4, "Álgebra": 5}

# Ahora mostraremos las asignaturas y sus créditos en la misma línea.
# Empezamos con imprimir los títulos
print("Asignaturas: Créditos")

# Ahora usamos un bucle "for" para recorrer cada par clave-valor
# (asignatura-crédito) en el diccionario 'asignaturas'. Usamos 'in' para iterar
# sobre el diccionario, y con 'items()' obtenemos las asignaturas y sus
# créditos. Usamos la f-string para mostrar cada asignatura y su
# crédito (variables) de lado a lado

for asignatura, credito in asignaturas.items():
    print(f"{asignatura}: {credito}")
# 3) Calculamos el total de créditos mediante la función sum (sumando valores)
# y procedemos a imprimir
total_creditos = sum(asignaturas.values())
print("\nEl total de créditos es:", total_creditos)

# 4) Ahora eliminamos el elemento "Álgebra" mediante .pop, y asignamos el valor
# 2 a Trigonometría
asignaturas.pop("Álgebra")
asignaturas["Trigonometría"] = 2

# Procedemos a mostrar el resultado final
print("\nAsignaturas: Créditos")
for asignatura, credito in asignaturas.items():
    print(f"{asignatura}: {credito}")
````

### Ejercicio 6

Cree un programa que calcule el sueldo bruto y neto de una persona, siendo así, en euros, lo que puede ganar en un mes de trabajo. Para ello pide esta (y sólo esta) información por teclado:

- Nombre de la persona
- Horas trabajadas
- Precio por hora

A partir de los anteriores datos debes hacer los cálculos necesarios para tener una salida por pantalla con el siguiente formato:

`" {nombre de la persona}  tiene un sueldo bruto de {sueldo bruto} euros y un sueldo neto de {sueldo neto} euros al mes" `


**NOTA 1:** Los valores del sueldo bruto y neto deben mostrarse solamente con un decimal.

**NOTA 2:** En el caso del sueldo neto solamente descontaremos el 15% sobre el sueldo bruto.

**(1.5 puntos)** $\color{orange}{\text{EG}}$

````python
# Respuesta

# 1. Solicitamos los datos mediante input. Usamos float para simplificar y
# permitir decimales ya desde el input, aunque se puede proceder con int() sin
# afectar al resultado final
nombre_de_la_persona = input("Introduce el nombre de la persona: ")
horas_trabajadas = float(input("Introduce las horas trabajadas: "))
precio_por_hora = float(input("Introduce el precio por hora: "))

# 2. Calculamos el sueldo bruto y neto
sueldo_bruto = horas_trabajadas * precio_por_hora
sueldo_neto = sueldo_bruto * (1 - 0.15)

# 3. Mostramos la salida por pantalla con el formato pedido. Usamos la f-string
# para incluir variables, dónde a la vez la .1f sirve para que se muestre el
# resultado formateado con un solo decimal
print(f"{nombre_de_la_persona} tiene un sueldo bruto de {sueldo_bruto:.1f} euros y un sueldo neto de {sueldo_neto:.1f} euros al mes")
````

### Ejercicio 7

Escribir un programa que pida al usuario que introduzca su nombre y su número telefónico.

El resultado del programa deberá ser una frase que constará de lo siguiente:

1. Mostrar el nombre del usuario en mayúsculas,

2. La cadena de texto: "tiene el siguiente número de teléfono", la cual deberá estar invertida.

3. El número de teléfono extrayendo los 3 primeros números.

Se ha de utilizar **únicamente slicing y las variables iniciales para declarar el nombre, el número y la cadena de caracteres**, es decir, sin utilizar variables auxiliares.

**(1.5 puntos)** $\color{green}{\text{NM}}$

````python
# Respuesta

# Pedimos insertar el nombre y el número
nombre = input("Introduce tu nombre: ")
telefono = input("Introduce tu número de teléfono: ")
mensaje = "tiene el siguiente número de teléfono"

# Mostramos el resultado como se pide directamente para no tener que usar
# variables auxiliares. Usamos .upper para mayúscula. Se han imprimido los
# dígitos tras 3 usando el slicing [3:] (aquí omitimos los 3 primeros dígitos).
# Usamos el slicing [::-1] sin especificar el inicio y fin (i.e., vacios)
# ya que cogemos toda la secuencia del mensaje. Además, usamos el -1
# para invertir el orden
print(f"{nombre.upper()} {mensaje[::-1]} {telefono[3:]}")
````
