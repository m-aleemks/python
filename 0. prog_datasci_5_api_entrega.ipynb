En la teoría hemos visto que existen 4 tipos básicos de respuestas, según el estado de la petición. A partir de esta información, crea un diccionario que tenga como llave (key) un ejemplo de cada tipo (status code: 2xx, 3xx, 4xx y 5xx), y, como valor (value) un mensaje asociado a la descripción del estado. A continuación, crea una función que, dada una URL, muestre por pantalla el status code con su mensaje asociado (a partir del diccionario creado previamente). Por ejemplo:

`[INFO] 200: Petición correcta`

Si el estado de la petición es 200, muestra también los 1000 caracteres del contenido.

Para comprobar que el código funciona correctamente, usa una URL de ejemplo de cada tipo.

**(1.5 puntos)**

$\color{green}{\text{NM}}$

```python
# Respuesta
import requests

# Diccionario de códigos de estado con descripciones
status_messages = {200: "Petición correcta",
                   301: "Redirección permanente",
                   404: "Recurso no encontrado",
                   500: "Error interno del servidor"}

def proces_url(url):
    """
    La función realiza una solicitud al (argumento) url, muestra el
    código de estado y su mensaje asociado. Si el código es 200, muestra los
    primeros 1000 caracteres del contenido.
    """
    response = requests.get(url)
    status_code = response.status_code

    # Obtenemos el mensaje asociado al código de estado y lo imprimimos
    mensaje = status_messages.get(status_code, "Estado no reconocido")
    print(f"[INFO] {status_code}: {mensaje}")

    # Mostramos el contenido si el código es 200
    if status_code == 200:
        print("\Primeros 1000 caracteres:\n")
        print(response.text[:1000])  # Mostramos hasta 1000 caracteres

# Probamos con URLs de ejemplo
urls = {"200": "https://jsonplaceholder.typicode.com/posts/1",
        "301": "http://github.com",
        "404": "https://jsonplaceholder.typicode.com/nonexistent",
        "500": "https://httpstat.us/500"}

# Procesamos cada URL y mostramos los resultados
for key, url in urls.items():
    print(f"\nProcesando URL (ejemplo {key}): {url}")
    proces_url(url)
```

```python
# Respuesta
# Importamos requests, random, y time.
# requests se utiliza para hacer peticiones HTTP en Python.
# Permite interactuar con APIs, como la de OTDB, enviando
# solicitudes y recibiendo respuestas en formatos como JSON.
import requests

# Usamos random para generar números aleatorios y hacer selecciones aleatorias.
# En este caso, se utiliza para elegir una categoría de preguntas
# aleatoriamente si no se especifica una categoría.
import random

# time ofrece funciones para trabajar con el tiempo. En este caso, se usa
# para introducir un retraso (con time.sleep()) entre las peticiones, lo cual
# es útil para evitar errores como el 429 (demasiadas solicitudes) en las APIs.
import time

# a) Petición a la API para obtener una pregunta aleatoria.
# La función get_random_question interactua con la API de Open Trivia Database
# para obtener una pregunta de trivia aleatoria.
# El parámetro amount=1 en la url indica que se quiere obtener solo 1 pregunta.
# Utiliza el módulo requests para realizar una petición HTTP de tipo GET a la
# URL especificada. Se utiliza requests para realizar una petición HTTP tipo
# GET a la URL dada. La respuesta se almacena en la variable response y se
# imprime si la petición es correcta, i.e., if response.status_code == 200.
# Usamos .json() para convertir el contenido del cuerpo de la respuesta
# de JSON a un diccionario, facilitando su manipulación.
def get_random_question():
    url = "https://opentdb.com/api.php?amount=1"
    response = requests.get(url)
    print(f"Estado de la petición: {response.status_code}")
    if response.status_code == 200:
        data = response.json()
        print("Cuerpo de la respuesta:")
        print(data)
        return data

# b) Mostramos la pregunta obtenida con sus respuestas correctas e incorrectas
def display_question(data):
    question_data = data['results'][0]  # Extraemos la 1a pregunta del resultado
    question = question_data['question']  # Extraemos el texto de la pregunta
    correct_answer = question_data['correct_answer']  # Extraemos respuesta corr
    incorrect_answers = question_data['incorrect_answers']  # Extraemos las
                                                            # respuestas incorr
    print("\nPregunta:")  # Indicamos que mostramos la pregunta
    print(f"- {question}")  # Mostramos el texto de la pregunta
    print(f"- Respuesta correcta: {correct_answer}")  # La respuesta correcta
    print(f"- Respuestas incorrectas: {incorrect_answers}")  # Las respuestas
                                                             # incorrectas

# c) Obtenemos las categorías disponibles
def get_categories():
    url = "https://opentdb.com/api_category.php"  # URL para obtener categorías
    response = requests.get(url)  # Realiza la petición a la API
    if response.status_code == 200:  # Comprueba si la respuesta fue exitosa
        # Obtiene el listado de categorías:
        categories = response.json().get("trivia_categories", [])
        # Devuelve un diccionario con nombre e ID de cada categoría:
        return {cat['name']: cat['id'] for cat in categories}
    return {}

# d) get_trivial_questions obtiene las preguntas trivias
def get_trivial_questions(amount=1, typ='multiple', difficulty='easy',
                          category=None):
    valid_difficulties = ['easy', 'medium', 'hard']
    valid_types = ['multiple', 'boolean']

    categories = get_categories()
    # Si no se proporciona una categoría, seleccionamos una aleatoriamente
    if category is None:
        category = random.choice(list(categories.values()))
    # Si la categoría es 'Books', la corregimos a 'Entertainment: Books'
    elif category == 'Books':
        category = 'Entertainment: Books'

    # Construimos la URL de la API
    url = f"https://opentdb.com/api.php?amount={amount}&category={category}" \
          f"&difficulty={difficulty}&type={typ}"

    print(f"URL generada: {url}")  # Imprimimos la URL generada para verificarla

    # Intentamos realizar la petición, con reintentos en caso de error 429
    retries = 3  # Número máximo de reintentos
    for attempt in range(retries):
        response = requests.get(url)

        # Si la respuesta es 200 (ok), procesamos los datos
        if response.status_code == 200:
            data = response.json()
            if data['response_code'] == 0:
                return data['results']
            elif data['response_code'] == 1:
                print("Error, no hay suficientes preguntas en esta " \
                      "categoría. Intentando con otra categoría...")
                return get_trivial_questions(amount, typ, difficulty)

        # Si el código es 429 (demasiadas peticiones), esperamos un tiempo
        # antes de reintentar
        elif response.status_code == 429:
            print("Error, límite de peticiones alcanzado. Intentando " \
                  "nuevamente tras 10s.")
            time.sleep(10)  # Esperamos 10 segundos para intentar de nuevo
        else:
            print(f"Error, no se pudo conectar con la API. Código de "
                  f"respuesta: {response.status_code}")
            break  # Salimos del bucle si la respuesta no es 200 ni 429

    return None


# Llamamos a las funciones
# a)
print("a) Pregunta aleatoria")
random_question = get_random_question()

# b)
print("\nb) Detalle de la pregunta")
display_question(random_question)

# c)
print("\nc) Categorías disponibles:")
categories = get_categories()
print(categories)

# d)
print("\nd.1) Solicitamos 2 preguntas sobre libros con dificultad alta")
questions = get_trivial_questions(amount=2, typ='multiple', difficulty='hard',
                                  category='Books')
if questions:
    for q in questions:
        print(q)

print("\nd.2) Llamamos con valores por defecto")
questions = get_trivial_questions()
if questions:
    for q in questions:
        print(q)

print("\nd.3) Dificultad inválida")
questions = get_trivial_questions(difficulty='dificil') # 'dificil' no es válido
if questions is None:
    print("Error: dificultad inválida.")
```

```python
# Respuesta
# Importamos requests y pandas como pd.
import requests
import pandas as pd

# Función auxiliar para manejar las peticiones a la API
def get_data_from_api(url):
    # Realizamos la petición a la API y devolvemos la respuesta en formato
    # JSON y los encabezados
    response = requests.get(url)
    return response.json(), response.headers

# a) Información del personaje Jon Snow
def get_jon_snow_info():
    # Establecemos la URL para obtener los datos del personaje
    # Jon Snow (posición 583)
    url = "https://anapioficeandfire.com/api/characters/583"
    # Extraemos los datos y los encabezados devueltos por la función
    # get_data_from_api(url) en data y _. Nota: el uso de _ como variable
    # indica que se ignora el valor asignado, pero, no es obligatorio usarlo
    # para ignorar un valor.
    data, _ = get_data_from_api(url)
    # Si hemos obtenido datos, los mostramos
    if data:
        # Imprimimos la información del personaje Jon Snow
        for key, value in data.items():
            print(f"- {key}: {value}")
    return data

# b) Número de páginas en la API de personajes
def get_number_of_pages():
    # Establecemos la URL para obtener los personajes
    url = "https://anapioficeandfire.com/api/characters"
    # Extraemos los encabezados de la respuesta
    _, headers = get_data_from_api(url)
    # Buscamos el campo 'Link' en los encabezados que contiene
    # la información de paginación
    link_header = headers.get('Link', '')
    if 'rel="last"' in link_header:
        # Mediante la línea siguiente extraemos la URL de la última página de
        # la API de la cabecera Link. Primero, la línea divide la cadena
        # link_header por comas, separando las URLs de las páginas. Luego,
        # selecciona la segunda parte (relacionada con la última página),
        # que contiene la URL y el identificador rel="last". A continuación,
        # se divide esa cadena por el punto y coma (;) y se toma la primera
        # parte, que es solo la URL. Finalmente, se eliminan los caracteres
        # angulares < y >, dejando solo la URL de la última página.
        last_page_url = link_header.split(',')[1].split(";")[0].strip("<>")
        # En la siguiente línea se toma la URL de la última página almacenada
        # en last_page_url. Luego se utiliza .split("=") para dividir la URL
        # en partes separadas por el signo =. Esto da como resultado una lista
        # en la que la última parte (tras el =) es el número total de
        # páginas. Se utiliza [-1] para obtener el último elemento de esa
        # lista, que corresponde al número de páginas. Finalmente,
        # se convierte ese número de páginas de tipo str a int usando int().
        num_pages = int(last_page_url.split("=")[-1])
        # Imprimimos el número de páginas encontrado
        return num_pages
    return 0

# c1) Lista de nombres y culturas de personajes
def get_names_and_cultures(num_pages):
    # Establecemos la URL base para obtener los personajes
    base_url = "https://anapioficeandfire.com/api/characters"
    # Creamos las listas vacías para los nombres y culturas
    names = []
    cultures = []
    # Recorremos todas las páginas
    for page in range(1, num_pages + 1):
        # Realizamos la petición para cada página de personajes
        data, _ = get_data_from_api(f"{base_url}?page={page}&pageSize=10")

        # Si obtenemos datos, procesamos los personajes
        if data:
            for character in data:
                # Extraemos el nombre y la cultura de cada personaje
                names.append(character.get('name', 'Unknown'))
                cultures.append(character.get('culture', 'Unknown'))

    # Imprimimos que hemos extraído los datos de nombres y culturas
    print("\nLista de nombres y culturas extraída.")
    return names, cultures

# c2) Lista de lealtades y libros de personajes
def get_allegiances_and_books(num_pages):
    # Establecemos la URL base para obtener los personajes
    base_url = "https://anapioficeandfire.com/api/characters"

    # Creamos las listas vacías para las lealtades y libros
    allegiances = []
    books = []

    # Recorremos todas las páginas
    for page in range(1, num_pages + 1):
        # Realizamos la petición para cada página de personajes
        data, _ = get_data_from_api(f"{base_url}?page={page}&pageSize=10")

        # Si obtenemos datos, procesamos los personajes
        if data:
            for character in data:
                # Hacemos una nueva petición para cada URL de
                # lealtad y obtenemos el nombre
                allegiance_names = []
                for url in character.get('allegiances', []):
                    allegiance_data, _ = get_data_from_api(url)
                    if allegiance_data:
                        allegiance_names.append(allegiance_data.get('name',
                                                                    'Unknown'))
                    else:
                        allegiance_names.append('Error al obtener lealtad.')
                allegiances.append(allegiance_names)

                # Libros: Hacemos una nueva petición para cada URL de libro
                # y obtenemos el nombre
                book_names = []
                for url in character.get('books', []):
                    book_data, _ = get_data_from_api(url)
                    if book_data:
                        book_names.append(book_data.get('name', 'Unknown'))
                    else:
                        book_names.append('Error al obtener libro')
                books.append(book_names)

    # Imprimimos que hemos extraído las lealtades y libros
    print("\nLista de lealtades y libros extraída.")
    return allegiances, books


# d) Crear DataFrame y filtrar por "Stark"
def create_dataframe(names, cultures, allegiances, books):
    # Creamos un diccionario con los datos de los personajes
    data = {"Name": names,
            "Culture": cultures,
            "Allegiances": allegiances,
            "Books": books}
    # Creamos un DataFrame a partir del diccionario
    df = pd.DataFrame(data)
    # Creamos una máscara booleana que identifica las filas que contienen
    # 'Stark' en la columna "Name"
    mask = df["Name"].str.contains("Stark", case=False, na=False)
    # Aplicamos la máscara al DataFrame para filtrar solo las filas que
    # contienen 'Stark' en el nombre
    stark_df = df[mask]

    # Imprimimos las primeras 5 filas del DataFrame filtrado
    print("\nDataFrame de personajes filtrado por 'Stark':")
    print(stark_df.head(5))
    return df

# Llamamos a las funciones y imprimimos los resultados
# a)
print("\na) Información de Jon Snow")
jon_snow_info = get_jon_snow_info()
print(jon_snow_info)
# b)
print("\nb) Número de páginas de personajes")
num_pages = get_number_of_pages()
print(f"Número total de páginas: {num_pages}")
# c.1)
print("\nc.1) Nombres y culturas de personajes")
names, cultures = get_names_and_cultures(num_pages)
print(f"Primeros 5 nombres: {names[:5]}")
print(f"Primeras 5 culturas: {cultures[:5]}")
# c.2)
print("\nc.2) Lealtades y libros de personajes")
allegiances, books = get_allegiances_and_books(num_pages)
print(f"Primeras 5 lealtades: {allegiances[:5]}")
print(f"Primeros 5 libros: {books[:5]}")
# d)
print("\nd) Crear DataFrame y filtrar por 'Stark'")
df = create_dataframe(names, cultures, allegiances, books)
```
