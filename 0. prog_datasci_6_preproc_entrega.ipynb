a) Observa la variable `Duración`. Extrae la duración total en minutos.
Convierte a formato numérico para facilitar análisis.

Muestra las últimas 260 filas comparando la columna original y la columna procesada.


Nota, considera el uso de [RegEx](https://www.w3schools.com/python/python_regex.asp)


**(1,5 punto)**   <span style="fuente-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span>  

```python
# Respuesta
import pandas as pd
import re
import numpy as np
from google.colab import drive
drive.mount('/content/drive')
ruta_archivo = "/content/drive/MyDrive/Colab_Notebooks" \
               "/prog_datasci_6/data/recetasdelaabuela.csv"
data = pd.read_csv(ruta_archivo)

# Normalizamos columnas como anteriormente
data.columns = (
    data.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace(r"[^a-zA-Z0-9_]", "", regex=True))

# Inspeccionamos valores únicos de la columna 'duracion'
print("Valores únicos de la columna duracion:")
print(data["duracion"].unique())


def convert_duracion_a_min(duracion):
    if pd.isnull(duracion):
        return np.nan
    # Normalización básica por si hay espacios adicionales
    duracion = duracion.strip().lower()
    # Extraemos horas y minutos mediante RegEx
    horas = re.search(r"(\d+)\s*h", duracion)
    # (\d+) busca uno o más dígitos consecutivos (\d+).
    # Los paréntesis () capturan estos dígitos para que puedan extraerse más
    # adelante como un grupo.
    # \s* permite la presencia de cero o más espacios en blanco (\s), e.g.,
    # "45m" y "45 m" serán válidos.
    # [mM] busca la letra "m", ya sea en minúscula (m) o mayúscula (M),
    # asegurando capturar todas la m.
    # Devuelve un objeto Match si encuentra una coincidencia; de lo contrario,
    # devuelve None.
    minutos = re.search(r"(\d+)\s*[mM]", duracion)
    total_minutos = 0
    if horas:
      # horas.group(1) captura el número de horas que ha encontrado la
      # expresión regular, donde group(1) hace referencia al primer grupo de la
      # expresión regular (el número de horas).
      # Así, si e.g. "duracion = "2 h 30 m"", group(1) extrae el primer grupo
      # de la coincidencia. En este caso, el grupo es el número de horas,
      # que es "2". Esto es lo que hace group(1).
      # "int(horas.group(1)) * 60" convierte las horas a minutos mediante * 60.
      # total_minutos acumula el valor calculado, i.e., a los minutos
      # acumulados hasta ese momento, se les añade la cantidad de minutos
      # correspondientes a las horas extraídas.
        total_minutos += int(horas.group(1)) * 60 # Convierte horas a minutos
    if minutos:
        total_minutos += int(minutos.group(1)) # Agrega minutos
    return total_minutos

# Aplicamos la función a la columna duracion
duracion_min = data["duracion"].apply(convert_duracion_a_min)
data["duracion_min"] = duracion_min
# Mostrar últimas 260 filas comparando ambas columnas
print("\nÚltimas 260 filas (duracion original y procesada):")
print(data[["duracion", "duracion_min"]].tail(260))
```

b) Observa la variable `Categoría`. Modifica la variable de forma que solo aparezca la primera categoría de la cada lista.

Rellena los valores faltantes con `NaN`.

**(0,5 puntos)**  <span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 2px; ">NM</span>

```python
# Respuesta
# Función para procesar valores de categoría
def procesar_categoria(valor):
    """
    Esta función procesa los valores de la columna 'categoria' de un dataset.
    Los valores pueden estar en formato de lista (e.g. ['Recetas mexicana',
    'Recetas peruana']) o ser cadenas simples (e.g. ['Recetas mexicanas']).
    La función realiza los siguientes pasos:
    - Si el valor es NaN o una lista vacía, devuelve NaN.
    - Si el valor es una lista no vacía, elimina los corchetes y las comillas,
      y procesa el valor.
    - Si el valor contiene múltiples categorías separadas por comas,
      toma solo la primera.
    valor (str): valor de la categoría a procesar. Devuelve str o np.nan, i.e.,
    la 1r categoría procesada o NaN si no es válido.

    Caso 1: si el valor es NaN o una lista vacía "[]", devolvemos NaN.
    Verificamos si el valor es NaN o si es una lista vacía "[]".
    """
    if pd.isnull(valor) or valor.strip() == "[]":
        return np.nan

    """
    Caso 2: limpiamos y procesamos categorías válidas.
    Eliminamos los corchetes ("[]") y las comillas simples (') de la cadena.
    Además, aplicamos el método strip() para quitar espacios en blanco extra
    alrededor del valor.
    """
    valor_limpio = valor.strip("[]").replace("'", "").strip()
    # Si el valor limpio queda vacío, devolvemos NaN.
    # Esto es útil para manejar casos donde el valor es una lista vacía o una
    # cadena con solo espacios.
    if not valor_limpio:
        return np.nan

    """
    Caso 3: tomamos la primera categoría si hay múltiples. Si el valor contiene
    varias categorías separadas por comas, tomamos solo la primera.
    Esto es útil cuando la categoría contiene más de una opción, e.g.,
    "Recetas mexicanas, Recetas peruanas".
    """
    return valor_limpio.split(",")[0].strip()

# Aplicamos la función a la columna "categoria".
# Utilizamos la función 'apply' para aplicar 'procesar_categoria' a cada
# elemento de la columna 'categoria'.
categoria_procesada = data["categoria"].apply(procesar_categoria)
data["categoria_procesada"] = categoria_procesada

# Verificamos resultados: valores únicos procesados.
# Mostramos los valores únicos de la columna "categoria_procesada".
print("Valores únicos en categoría procesada:")
print(data["categoria_procesada"].unique())

# Mostramos un subconjunto de datos transformados, incluyendo celdas vacías.
# Se selecciona el rango de filas 971:985.
print("\nEjemplo de transformación (categoría original y procesada):")
print(data[["categoria", "categoria_procesada"]].iloc[971:985])
```

c) Observa la variable `Valoración y votos`. Divide la variable en dos columnas:

- Valoración: Un valor numérico flotante (e.g., 4.5).
- Votos: Un valor entero que representa la cantidad de votos (e.g., 200).

Rellena los valores faltantes con `NaN`.

Comprueba el código mostrando por pantalla los indices `19015:19029` para las columnas `valoracionyvotos`, `valoracion` y `votos`.


**(0,5 puntos)**  <span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 2px; ">NM</span>

```python
# Respuesta
import pandas as pd
import numpy as np
import re

def procesar_valoracion_y_votos(valoracion_votos):
    """
    Función para procesar la columna 'valoracion_y_votos'.
    Esta función extrae la valoración y los votos de una cadena de texto.
    Devuelve NaN si no se extrae la información correctamente.

    Caso 1: si el valor es NaN o está vacío, devolvemos NaN para ambas columnas.
    Verificamos si el valor es NaN o si el campo está vacío (NaN).
    En cualquiera de estos casos, devolvemos NaN en ambas columnas.
    """
    if pd.isnull(valoracion_votos):
        return np.nan, np.nan

    """
    Caso 2: extraemos valoración y votos utilizando regex.
    Usamos 're.findall' para buscar un patrón que extraiga primero los valores
    numéricos de la valoración
    (con decimales permitidos) y después el número de votos (un número entero).
    El patrón '([\d.]+).*?(\d+)' busca un número flotante (valoración) seguido
    de un número entero (votos).

    [\d.] indica que estamos buscando cualquier dígito (\d, que equivale a
    [0-9]) o un punto (.), ya que un número decimal puede tener puntos decimales.
    + significa "uno o más" de los caracteres anteriores, i.e., buscará una
    secuencia de dígitos o puntos. Esto captura la valoración (e.g., 4.5 o 3).
    En ").*?" ) cierra el primer grupo de captura (la valoración).
    En ".*?" el punto (.) significa "cualquier carácter", y el *? indica
    "cero o más veces de manera no codiciosa" (es decir, el patrón buscará la
    menor cantidad de caracteres posibles que coincidan). Este fragmento ayuda
    a descartar cualquier texto entre la valoración y los votos,
    hasta llegar a la siguiente parte relevante.
    (\d+) busca un número entero (sin decimales).
    \d representa cualquier dígito, y el
    + significa "uno o más" dígitos. Este fragmento captura los votos.
    """
    resultado = re.findall(r"([\d.]+).*?(\d+)", valoracion_votos)

    """
    Caso 3: si encontramos el patrón, devolvemos la valoración como un número
    flotante y los votos como un entero.
    Si no encontramos el patrón esperado, devolvemos NaN para ambas columnas.
    resultado[0][0] corresponde a la valoración (el 1r valor numérico
    extraído de la cadena).
    resultado[0][1] corresponde a los votos (el 2n valor
    extraído de la cadena).
    E.g., si el resultado[0][0] es 0 y resultado[0][1] es un valor mayor que 0,
    significa que sólo se extrajo el número de votos correctamente,
    pero la valoración no se pudo extraer, lo que también generará NaN para la
    valoración y el valor correcto para los votos.
    """
    if resultado:
        valoracion = resultado[0][0]
        votos = resultado[0][1]
        return float(valoracion), int(votos)

    return np.nan, np.nan

# Aplicamos la función 'procesar_valoracion_y_votos' a cada fila de la columna
# 'valoracion_y_votos'.
# 'apply' aplica la función a cada valor en la columna, y 'pd.Series' convierte
# el resultado en dos columnas separadas.
valoracion_votos_procesados = data["valoracion_y_votos"].apply(
    lambda x: pd.Series(procesar_valoracion_y_votos(x)))

# Asignamos los resultados de las dos columnas (valoración y votos) a nuevas
# columnas en el DataFrame.
# 'valoracion' recibirá el 1r valor de la serie procesada, y 'votos' el 2n.
data["valoracion"] = valoracion_votos_procesados[0]
data["votos"] = valoracion_votos_procesados[1]

# Verificamos los resultados mostrando el rango 19015:19029.
print("\nÍndices 19015:19029 para las columnas procesadas:")
print(data.loc[19015:19029, ["valoracion_y_votos", "valoracion", "votos"]])
```

d) Observa la variable `Comensales`. Verificar si los valores son numéricos o si están en texto (e.g., "4 personas"). Convertir los valores a un formato numérico en una nueva variable `comensales_num`.

Rellena los valores faltantes con `NaN`.

Comprueba el código mostrando por pantalla los indices `19015:19029` para la columna sin procesar y la procesada.

**(1 punto)**  <span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 2px; ">NM</span>

```python
# Respuesta
import pandas as pd
import numpy as np
import re
# Función para procesar la columna 'comensales'
# Esta función toma un valor de la columna 'comensales' y extrae el número de
# comensales. Si el valor es NaN o no contiene un número válido, devolvemos NaN
def procesar_comensales(comensales):
    """
    Procesa los valores de la columna 'comensales'. Si el valor es un número
    válido, devuelve el número de comensales como un entero. Si no es posible
    extraer un número, devuelve NaN.
    # Caso 1: si el valor es NaN, devolvemos NaN.
    # Verificamos si el valor es NaN. En ese caso, devolvemos NaN.
    """
    if pd.isnull(comensales):
        return np.nan
    """
    Caso 2: extraemos el número de comensales usando una expresión regular
    Utilizamos regex para encontrar el primer número dentro del texto.
    """
    resultado = re.search(r"(\d+)", comensales)
    """
    Caso 3: si encontramos un número, lo devolvemos como entero.
    Si encontramos un número, lo convertimos a entero y lo devolvemos.
    """
    if resultado:
        comensales_num = int(resultado.group(1)) # Asignamos el nº de comensales
        return comensales_num

    # Si no encontramos un número, devolvemos NaN.
    return np.nan

# Aplicamos la función a la columna 'comensales'
# Usamos el método 'apply' para aplicar la función 'procesar_comensales' a cada
# fila de la columna 'comensales'
comensales_procesados = data["comensales"].apply(procesar_comensales)

# Creamos la nueva columna 'comensales_num'
# Asignamos los resultados procesados a la nueva columna 'comensales_num' del df
data["comensales_num"] = comensales_procesados

# Mostrar los índices solicitados
# Mostramos los valores de las columnas 'comensales' original y
# 'comensales_num' procesada para los índices de 19015 a 19029
print("\nÍndices 19015:19029 para la columna 'Comensales' procesada:")
print(data.loc[19015:19029, ["comensales", "comensales_num"]])
```

```python
# Respuesta
import pandas as pd
from collections import Counter
def clasificar_y_contar(termino):
    """
    Clasifica y divide los términos en la columna 'categoria_2'. Si el valor es
    NaN, devuelve una lista vacía. Si no, separa los términos por comas y los
    limpia de espacios adicionales. Devuelve una lista de términos.
    Si el valor es NaN, devuelve una lista vacía.
    # Caso 1: si el valor es NaN, devolvemos una lista vacía.
    """
    if pd.isnull(termino):
        return []
    """
    # Caso 2: dividimos el término por comas y eliminamos espacios en blanco
    # adicionales. Utilizamos split para dividir y strip para limpiar.
    """
    return [t.strip() for t in termino.split(",")]
# Usamos 'apply' para aplicar la función 'clasificar_y_contar' a
# cada fila de la columna 'categoria_2'. Luego usamos 'explode' para separar
# cada término en una fila individual.
terminos = data["categoria_2"].dropna().apply(clasificar_y_contar).explode()
# Usamos 'Counter' para contar las veces que aparece cada
# término en la lista 'terminos'.
conteo_terminos = Counter(terminos)

# Iteramos sobre el diccionario de conteo y clasificamos cada término
# en una categoría.
clasificacion = []
for termino, total in conteo_terminos.items():
    # Caso 1: si el término empieza con "sin", lo clasificamos como "sin".
    if termino.lower().startswith("sin"):
        categoria = "sin"
    # Caso 2: si el término empieza con "bajo", lo clasificamos como "bajo".
    elif termino.lower().startswith("bajo"):
        categoria = "bajo"
    # Caso 3: si el término empieza con "alto", lo clasificamos como "alto".
    elif termino.lower().startswith("alto"):
        categoria = "alto"
    # Caso 4: si no empieza con "sin", "bajo" ni "alto", lo clasificamos
    # como "otros".
    else:
        categoria = "otros"
    clasificacion.append({"Categoria": categoria, "Termino": termino,
                          "Total": total})
# Creamos un DataFrame con la clasificación y lo ordenamos por la frecuencia
# de los términos.
df_clasificacion = pd.DataFrame(clasificacion).sort_values(by="Total",
                                                           ascending=False)
# Mostramos los primeros 10 términos clasificados por su frecuencia.
print("\nTabla de términos clasificados:")
print(df_clasificacion.head(10))
```

b)  Seleccionar tres políticas para la imputación de datos faltantes mediante el método `SimpleImputer` del paquete [scikit-learn](https://scikit-learn.org/stable/).
Calcular el rendimiento de cada política de imputación para cada variable en términos de Root Mean Square Error.



<span style="fuente-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span> **(1,5 puntos )**

```python
# Respuesta
# Importamos las bibliotecas necesarias
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.metrics import mean_squared_error
from google.colab import drive
drive.mount('/content/drive')
ruta_diamonds = "/content/drive/MyDrive/Colab_Notebooks/prog_datasci_6/" \
                "data/diamonds.csv"
original_df = pd.read_csv(ruta_diamonds)

# Seleccionamos las columnas numéricas para la imputación
columns_to_modify = ["carat", "depth", "table", "price", "x", "y", "z"]
# Creamos una copia del dataset original para introducir valores faltantes
df = original_df.copy()

# Se genera un 20% de datos faltantes aleatorios en las columnas seleccionadas
# para simular un escenario donde los datos tienen valores ausentes.
# Para cada una de las tres políticas (mean, median, most_frequent),
# creamos un imputador y aplicamos esa estrategia a los datos faltantes.
# Para cada columna y cada estrategia, se calcula el Root Mean Square Error
# (RMSE) comparando los valores imputados con los valores reales del
# dataset original.
# Se presenta un DataFrame con los resultados de RMSE para cada estrategia y
# cada variable, lo que permite comparar la efectividad de las tres
# políticas de imputación.
np.random.seed(42)  # Usamos una semilla para reproducibilidad
for column in columns_to_modify:
    n_missing = int(0.2 * len(df[column]))  # Calculamos el 20% de filas
    # Seleccionamos índices aleatorios
    missing_indices = np.random.choice(df.index, n_missing, replace=False)
    # Asignamos NaN en los índices seleccionados
    df.loc[missing_indices, column] = np.nan

# Verificamos el porcentaje de valores faltantes por variable y los imprimimos
missing_percentage = df[columns_to_modify].isnull().mean() * 100
print("\nPorcentaje de valores perdidos por variable:")
print(missing_percentage)

# Definimos las estrategias de imputación a evaluar
strategies = ["mean", "median", "most_frequent"]
# Diccionario para almacenar los resultados de RMSE
rmse_results = {}
# Evaluamos cada estrategia de imputación. strategy es una variable de control
# en este caso, que en cada iteración toma uno de los valores de la lista
# strategies ("mean", "median", o "most_frequent").
for strategy in strategies:
    imputers = {}  # Diccionario para almacenar imputadores por columna
    rmse_results[strategy] = {}  # Diccionario para guardar RMSE por estrategia
    for column in columns_to_modify:
        # Creamos el imputador para la estrategia actual
        imputer = SimpleImputer(strategy=strategy)
        imputers[column] = imputer
        # Aplicamos la imputación a los datos con valores faltantes
        df_imputed = df.copy()  # Trabajamos con una copia del dataset con NaN
        df_imputed[column] = imputer.fit_transform(df[[column]])
        # Identificamos los índices donde realizamos imputación
        valid_indices = original_df[column].notnull() & df[column].isnull()
        # Calculamos el RMSE comparando los valores originales con los imputados
        # valid_indices identifica las filas donde se imputaron valores.
        # mean_squared_error calcula el error cuadrático medio entre los
        # valores reales y los imputados.
        # np.sqrt() toma la raíz cuadrada de ese valor para obtener el RMSE,
        # que es la métrica de rendimiento para evaluar qué tan bien funcionó
        # la imputación en esos índices seleccionados.
        rmse = np.sqrt(
            mean_squared_error(
                original_df.loc[valid_indices, column],  # Valores originales
                df_imputed.loc[valid_indices, column]))   # Valores imputados
        # Guardamos el RMSE en el diccionario de resultados
        # rmse_results[strategy] accede al subdiccionario de resultados para la
        # estrategia de imputación actual ("mean", "median", o "most_frequent")
        # [column] selecciona la clave dentro de ese subdiccionario
        # correspondiente a la columna específica (e.g., "carat").
        # = rmse asigna el valor del RMSE calculado a esa posición
        # específica en el diccionario.
        rmse_results[strategy][column] = rmse
# Convertimos los resultados de RMSE a un DataFrame para mejor visualización
rmse_df = pd.DataFrame(rmse_results)

# Mostramos los resultados finales
print("\nRMSE por estrategia y variable:")
print(rmse_df)
```

c) Imputar los datos con valores perdidos con la mejor política para cada variable y crear un dataframe llamado `df`. Compruebe que no hay valores perdidos para cada variable


**(1 punto)**  <span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 2px; ">NM</span>

```python
# Respuesta
# Utilizamos el df 'rmse_df' que contiene los resultados de RMSE para cada
# estrategia de imputación y columna. Seleccionamos la estrategia con el menor
# RMSE para cada columna (axis=1 se refiere a columnas).
best_policies = rmse_df.idxmin(axis=1)

# Creamos una copia del df original para no modificarlo directamente.
df_imputed = original_df.copy()
# Recorremos cada columna y su respectiva mejor estrategia de imputación
for column, best_strategy in best_policies.items():
    # Creamos un SimpleImputer para cada columna con la estrategia
    # seleccionada en best_strategy. El imputador se utiliza para reemplazar
    # los valores faltantes (NaN) con el valor calculado según
    # la estrategia ('mean', 'median', o 'most_frequent').
    best_imputer = SimpleImputer(strategy=best_strategy)
    # Aplicamos el imputador a la columna seleccionada.
    # fit_transform() es de la librería Scikit-learn, dónde
    # el método fit() se usa para ajustar el imputador a los datos.
    # En este caso, el imputador aprende de los valores no nulos en la columna,
    # y dependiendo de la estrategia seleccionada,
    # calcula el valor que se usará para reemplazar los valores faltantes.
    # El método transform() se usa para aplicar la transformación
    # (en este caso, la imputación) a los datos, i.e., reemplaza los valores
    # faltantes con los valores calculados durante el ajuste (e.g. la media).
    df_imputed[column] = best_imputer.fit_transform(df[[column]])

# Comprobamos que no queden valores perdidos en el df imputado.
# Si todo ha ido bien, el resultado de 'isnull().sum()' será 0 para
# todas las columnas.
print("Valores perdidos por variable después de la imputación:")
print(df_imputed.isnull().sum())
```
