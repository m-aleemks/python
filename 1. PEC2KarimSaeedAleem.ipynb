# Ejercicio 1

Queremos crear una función que compruebe si se puede realizar la mezcla entre ingredientes de una pizza. Los ingredientes que puede introducir se clasifican en dos tipos:

Principales:

- Tofu
- Bacon
- Embutido picante

Secundarios:

- Aceituna
- Champiñones
- Piña

La función recibirá dos parámetros:
- Una lista con 3 ingredientes
- un parámetro **opcional** que nos indique si queremos que la función muestre por pantalla la razón por la cual no se puede realizar la pizza (en el caso de que si se pueda, la función no deberá mostrar ningún mensaje de todas formas).

La función devolverá un dato de tipo booleano **(True / False)**.

Según la pizzeria, los únicos ingredientes que no se pueden poner en misma pizza son los siguientes:

- Bacon y Tofu excepto si se le añade aceituna
- Bacon, champiñones y Piña
- Embutido picante y aceituna excepto si se añade champiñones

Las combinaciones han de ser obligatoriamente de 3 ingredientes pero la única combinación de 2 ingredientes permitida es tofu con aceituna.

No se puede hacer una mezcla de más de 3 ingredientes ni menos de 3, a excepción de la que se ha comentado.

Como segunda parte de este ejercicio, **utilizando programación funcional**,
realiza lo siguiente:

- Cuáles de estas combinaciones de ingredientes se pueden realizar y cuáles no
- Mostrar la/s tupla/s de los ingredientes que se pueden realizar.

<pre>
\[ ["Bacon", "Embutido picante", "Piña"],["Aceituna", "Piña", "Champiñones"],["Tofu", "Bacon", "Aceituna", "Piña"] \]
</pre>

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 punto)**

```python
# Respuesta
# Los ingredientes principales y secundarios son:
# Principales = {"Tofu", "Bacon", "Embutido picante"}
# Secundarios = {"Aceituna", "Champiñones", "Piña"}


def es_pizza_valida(ingredientes):
    """
    La función comprueba si una combinación de ingredientes es válida para
    hacer una pizza, donde el argumento ingredientes es la lista de
    ingredientes de la pizza (máx. 3 elementos). La función devuelve un bool
    True si la combinación es válida, False en caso contrario.
    """

    # Verificamos la cantidad de ingredientes
    if len(ingredientes) not in (2, 3):
        return False

    # E.g., en la primera condicón prohibimos las combinaciones que contengan
    # "Bacon" y "Tofu", pero permitimos esa combinación si también incluye
    # "Aceituna" porque la condición "Aceituna" not in ingredientes"
    # sería falsa en tal caso.
    if ("Bacon" in ingredientes and "Tofu" in ingredientes and
       "Aceituna" not in ingredientes):
        return False

    if ("Bacon" in ingredientes and "Champiñones" in ingredientes and
       "Piña" in ingredientes):
        return False

    if ("Embutido picante" in ingredientes and "Aceituna" in ingredientes and
       "Champiñones" not in ingredientes):
        return False

    # Con esta condición prohibimos todas las combinaciones de 2 ingredientes
    # que incluyan "Tofu" y no incluyan "Aceituna". Pero si la lista contiene
    # únicamente "Tofu" y "Aceituna", no activa esta regla, permitiendo que
    # el código dé True.
    if (len(ingredientes) == 2 and "Tofu" in ingredientes and
       "Aceituna" not in ingredientes):
        return False

    return True


def analizar_comb(combinaciones):
    """
    La función analiza qué combinaciones de ingredientes son válidas. El
    argumento combinaciones es la lista de combinaciones de ingredientes.
    Devuelve un tuple de la lista de combinaciones válidas y no válidas.

    Más concretamente, empezamos creando listas vacías para convertir las
    listas a tuplas:
      comb_val almacena combinaciones que cumplen las reglas.
      comb_inval almacena las que no cumplen las reglas.
    El bucle for itera sobre cada combinación c en la lista de combinaciones.
    La condición if usa es_pizza_valida(c) para determinar si la combinación
    cumple las reglas:
      Si es válida, convierte c a una tupla con tuple(c) y la añade a
      comb_val.
      Si no es válida, convierte c a tupla y la añade a comb_inval.
    """
    comb_val = []
    comb_inval = []

    for c in combinaciones:
        if es_pizza_valida(c):
            comb_val.append(tuple(c))
        else:
            comb_inval.append(tuple(c))

    return comb_val, comb_inval


# Ejecutamos las funciones anteriores mediante las combinaciones dadas:
combinaciones = [["Bacon", "Embutido picante", "Piña"],
                 ["Aceituna", "Piña", "Champiñones"],
                 ["Tofu", "Bacon", "Aceituna", "Piña"]]

# Ahora analizamos combinaciones. La función analizar_comb devuelve
# una tupla de dos listas. Luego, Python automáticamente desempaqueta la
# tupla y asigna la primera lista a "validas" y la segunda lista a "invalidas".
validas, invalidas = analizar_comb(combinaciones)

# Imprimimos las combinaciones válidas
print("Combinaciones válidas:")
for c in validas:
    print(c)

# Imprimimos las combinaciones inválidas
print("\nCombinaciones inválidas:")
for c in invalidas:
    print(c)
```

Un equipo de científicos quiere diseñar diferentes funcionalidades para procesar información y extraer diferentes métricas utilizando la base de volcanes. Los datos los pueden visualizar en el excel volcanes.xlsx

  - Llamada "elevation_km": En donde se conviertan los valores de la columna "Elevation (m)" de metros a km

- 2.3 (0,5 puntos) crear una función que reciba el diccionario del apartado anterior como parámetro de entrada y que devuelva otro diccionario únicamente con los nombres y los valores de los cinco volcanes con más elevación ordenados de mayor a menor.

Recuerda mostrar los resultados por pantalla en cada subapartado

**Notas importantes:**

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 puntos)**

```python
# Respuesta
# Importamos el Drive, pandas y los datos.
from google.colab import drive
import pandas as pd
drive.mount('/content/drive')
data = pd.read_excel('/content/drive/My Drive/Colab_Notebooks/'
                     'Programación_DS/Activity_2/data/volcanes.xlsx')


# 2.1)
def agregar_columna_elevacion_km(data):
    """
    La función agrega una columna que convierte la elevación de metros a
    kilómetros y muestra el resultado. "data" es el DataFrame con los datos de
    los volcanes. La función devuelve el DataFrame actualizado.
    """
    # Creamos nueva columna en km dividiendo "elevation (m)" por mil e
    # imprimimos los resultados por pantalla.
    data["elevation_km"] = data["elevation (m)"] / 1000

    print("\n2.1.)\nDataFrame con la nueva columna 'elevation_km':")
    print(data.head())

    return data


# 2.2)
def create_dictionary_with_elevation(data):
    """
    La función devuelve un diccionario donde las claves son las regiones de los
    volcanes y los valores su elevación en metros. "data" contiene los datos de
    los volcanes. La función devuelve el diccionario con las regiones de los
    volcanes como claves y su elevación en metros como valores.

    Creamos el diccionario e imprimimos los resultados.
    zip() es una función que toma dos o más iterables, e.g. listas o,
    columnas, y las combina en tuplas. En este caso, combina cada valor de la
    columna volcano name con su correspondiente valor en elevation (m).

    dict() convierte las tuplas generadas por zip() en un diccionario.
    Así, cada "volcano name" se convierte en una clave y su "elevación (m)"
    en el valor correspondiente.
    """
    volcanes_dict = dict(zip(data["volcano name"], data["elevation (m)"]))

    print("\n2.2.)\nDiccionario con regiones y elevaciones:")
    print(volcanes_dict)

    return volcanes_dict


# 2.3)
def obtener_top_5_volcanes_elevacion(volcanes_dict):
    """
    La función recibe un diccionario con nombres de volcanes y sus elevaciones
    y devuelve otro diccionario con los 5 volcanes más altos ordenados de mayor
    a menor. volcanes_dict es el diccionario con nombres de volcanes como
    claves y su elevación como valores. La función devuelve un diccionario
    con los nombres y elevaciones de los cinco volcanes más altos.

    Obtenemos los top 5 e imprimimos los resultados. Usamos items() es un
    (método de diccionario) para obtener los pares clave-valor.

    Usamos sorted() para ordenar las tuplas de la lista de forma que
    elevación (valor) sea el criterio de orden. Por defecto, sorted() ordena
    las tuplas ascendentemente. Por ello, usamos "reverse=True" para ordenar
    en orden descendente.

    La función anónima "key=lambda x: x[1]: lambda x: x[1]" indica que la
    ordenación debe hacerse según el segundo elemento de la tupla, i.e.,
    la elevación del volcán.

    Tras usar "reverse=True" para ordenar descendentemente, usamos [:5] para
    seleccionar los primeros 5 elementos de la lista ordenada, que ahora serán
    los 5 volcanes más altos.

    Usamos dict() para convertir la lista de tuplas ordenadas nuevamente
    en un diccionario, y al final imprimimos y devolvemos los resultados.
    """
    top_5_volcanes = dict(sorted(volcanes_dict.items(), key=lambda x: x[1],
                                 reverse=True)[:5])

    print("\n2.3.)\nTop 5 volcanes con mayor elevación:")
    print(top_5_volcanes)

    return top_5_volcanes


# Ejecutamos el código
# Imprimimos columnas del archivo para confirmar los nombres
print("Columnas en el DataFrame cargado:")
print(data.columns.tolist())

# Normalizamos nombres de columnas y las imprimimos
data.columns = data.columns.str.strip().str.lower()

print("\nColumnas tras normalización:")
print(data.columns.tolist())

# Notas: los nombres de los parámetros en las funciones (data, volcanes_dict)
# son solo convenciones. Pueden ser diferentes a los nombres de las variables
# que usamos al llamar las funciones. Lo importante es que lo que se pase como
# argumento sea del tipo que la función espera, i.e., si la función espera un
# df, se pasa un df. Si espera un diccionario, se pasa un diccionario.
# Así, los nombres de las variables no tienen que coincidir con los nombres de
# los parámetros.

# 2.1: agregamos la columna 'elevation_km'
volcanes_df = agregar_columna_elevacion_km(data)

# 2.2: creamos un diccionario con elevaciones
volcanes_dict = create_dictionary_with_elevation(volcanes_df)

# 2.3: obtenemos el top 5 de los volcanes más altos
top_5_volcanes = obtener_top_5_volcanes_elevacion(volcanes_dict)
```

# Ejercicio 3

Se dispone de un diccionario `ventas` de un comercio, el cual dispone de los productos, que es la clave del diccionario, y la cantidad vendida (valor) a lo largo de un mes.

Por ejemplo:

ventas = {'agua': 1000, 'leche': 300, 'sal': 1000, 'azúcar': 350,'aceite': 312}

También se dispone de un diccionario *stock_ini*, tal que
los items son los nombres de productos (clave) y la cantidad
del producto (valor) que que el comercio tenía a primeros del mes.

Por ejemplo:

stock_ini = {'sal':1000, 'aceite':1000, 'leche':500, 'macarrones': 400, 'azúcar':500, 'colacao': 1600, 'agua': 1000}

Al final de mes se tiene que actualizar el stock de acuerdo
con el que se ha vendido.

Por los ejemplos anteriores el nuevo stock seria:

stock_final = {'sal': 0,
'aceite': 688,
'leche':200,
'macarrones': 400,
'azúcar':150,
'colacao': 1600,
'agua': 0}

Consideraremos que que nunca se ha vendido más cantidades de un producto de la que había en stock.

Diseña la función `inventario(ventas, stock_ini)` que calcule y muestre el stock final según la especificación y además que muestre un mensaje que indique que procuctos son los que ha de reponer.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 puntos)**

```python
# Respuesta
def inventario(ventas, stock_ini):
    """
    La función actualiza el stock de productos según las ventas y muestra
    los productos que necesitan reponer, donde ventas es el diccionario con los
    productos vendidos y sus cantidades y stock_ini (dict) es el diccionario
    con el stock inicial de los productos. La función devuelve dict, un
    diccionario con el stock final actualizado.

    Notas: cuando asignamos un diccionario a otra variable en Python, no
    estamos creando una copia, sinó una referencia al mismo objeto en memoria,
    i.e., si modificamos el contenido de la nueva variable, también estamos
    modificando el original, porque ambos apuntan al mismo lugar en memoria.
    Cuando usamos stock_ini.copy(), creamos un nuevo objeto en memoria que
    contiene los mismos datos, pero no está vinculado al original.
    Esto asegura que cualquier cambio en stock_final no afecte al stock_ini.
    """
    stock_final = stock_ini.copy()

    # Usamos el siguiente bucle for para recorrer cada producto y su cantidad
    # vendida en el diccionario ventas. En cada iteración:
    # producto (nombres) es la clave, y cantidad_vendida es el valor
    # (cantidades). Este bucle permite actualizar el stock en stock_final
    # restando la cantidad vendida de cada producto. En breve, usamos
    # "for producto, cantidad_vendida in ventas.items()" para recorrer el
    # diccionario ventas y obtener en cada iteración el nombre del producto y
    # su cantidad vendida. Usamos este bucle como referencia para hacer las
    # operaciones de actualización en el diccionario "stock_final".
    for producto, cantidad_vendida in ventas.items():
        # Restamos la cantidad vendida al stock inicial, sin permitir
        # valores negativos usando max ya que max toma el valor más grande
        # entre los presentes, y el 0 será siempre mayor que un negativo.
        stock_final[producto] = max((stock_final[producto] -
                                     cantidad_vendida, 0))

    # Identificamos los productos agotados que necesitan reponer.
    # Nota: es como si antes del for siguiente hubiese un = que nos permite
    # guardar en el primer producto los nombres de los productos con cantidad
    # 0. Obviamente, no es necesario escribirlo explícitamente.
    prod_a_reponer = [producto for producto, cantidad in
                      stock_final.items() if cantidad == 0]

    # Mostramos el stock final
    print("Stock final actualizado:")
    for producto, cantidad in stock_final.items():
        print(f"- {producto}: {cantidad}")

    # Mostramos los productos que necesitan reponer
    if prod_a_reponer:
        print("\nProductos que necesitan reponerse:")
        for producto in prod_a_reponer:
            print(f"- {producto}")
    else:
        print("\nNo hay productos que necesiten reponerse.")

    return stock_final


# Datos utilizados para ejecutar la función
ventas = {'agua': 1000, 'leche': 300, 'sal': 1000,
          'azúcar': 350, 'aceite': 312}
stock_ini = {'sal': 1000, 'aceite': 1000, 'leche': 500, 'macarrones': 400,
             'azúcar': 500, 'colacao': 1600, 'agua': 1000}

# Llamamos a la función con lo siguiente
stock_final = inventario(ventas, stock_ini)
```

# Ejercicio 4

Se pide realizar una función que recorra el excel nombres.xlsx:

La función que se solicita se va a llamar `random_asignation`, recibirá como entrada:

- El archivo de datos en formato *.xlsx*.

- Un número comprendido entre 1 y 245.

La salida será un diccionario donde:

- La clave será el número correspondiente a la posición, es decir, ANTONIO, que es el primer nombre, tendrá de clave un 1.

- Como valor, una tupla que contendrá el nombre y un número aleatorio comprendido entre 1 y 100 que deberéis crear.

Para acabar, con el número que se ha solicitado al principio, se va a crear esa cantidad de números aleatorios y corresponderán a las claves que se van a mostrar como salida final de la función. Es decir, si me han indicado un 3 al principio, he de generar 3 números aleatorios, en este caso me han aparecido: 1, 5 y 10. La función deberá devolver:

\[('ANTONIO', 20), ('DAVID', 7),('JOSE ANTONIO', 21)\]

Recordad que los números que aparecen al lado del nombre son aleatorios por lo que no os aparecerán los mismos números que en este ejemplo.


<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span> **(2.0 puntos)**

```python
# Respuesta
# Importamos las librerías random y pandan junto al drive. Usamos random para
# generar números aleatorios.
import random
import pandas as pd
from google.colab import drive

# Montamos el Drive y obtenemos la ruta del archivo
drive.mount('/content/drive')
df_nombres = pd.read_excel('/content/drive/My Drive/Colab_Notebooks/'
                           'Programación_DS/Activity_2/data/nombres.xlsx')


def random_asignation(df_nombres, num):
    """
    La función genera un diccionario que asigna a cada posición un nombre y un
    número aleatorio. Los parámetros son el df_nombres, (datos del archivo
    Excel con los nombres), y el int num, que es la cantidad de muestras random
    a devolver de 1 a 245. La función devuelve una lista de tuplas con los
    nombres seleccionados aleatoriamente y sus valores asignados.

    Nota: el ":" actúa como el separador entre la clave y el valor dentro de
    cada par clave-valor del diccionario. Entonces, Python entiende
    automáticamente que aquí estamos creando un diccionario cuando ve el ":"
    entre dos valores.

    Nota: la estructura básica de list comprehensions es
      "[new_element for element in iterable if condition]"   , dónde
    "new_element" es lo que queremos incluir en la nueva lista.
    "for element in iterable" itera sobre un iterable, e.g. una lista o rango.
    "if condition" filtra los elementos que cumplen una condición. Es opcional.
    """

    # df_nombres.iloc[:, 0] selecciona toda la primera columna del df.
    # tolist() convierte esa columna en una lista manejable para Python.
    nombres_lista = df_nombres.iloc[:, 0].tolist()

    # Creamos un diccionario con claves como posiciones y valores como nombres.
    # enumerate(nombres_lista) devuelve un iterador de pares índice-valor,
    # donde i es el índice (que empieza en 0) y nombre es el valor
    # correspondiente de la lista de nombres.
    # Usamos "i + 1: nombre" para que las claves del diccionario comiencen en 1
    # y no 0. El valor de cada clave es el nombre correspondiente de la lista.
    # Usamos random.randint para generar un número random entre 1 y 100 para i.
    nombres_dicc = {i + 1: (nombre, random.randint(1, 100))
                    for i, nombre in enumerate(nombres_lista)}

    # Generamos una lista de claves aleatorias que representan posiciones en
    # el diccionario. Usamos random.sample para generar la lista num de números
    # únicos seleccionados aleatoriamente mediante el rango
    # "range(1, len(nombres_lista) + 1)", que es el rango de claves válidas
    # del 1 hasta la cantidad total presente. Sumamos 1 para que el último
    # número (la clave máxima) sea igual al número total de nombres,
    # que en este caso son 245.
    claves_rand = random.sample(range(1, len(nombres_lista) + 1), num)

    # Inicializamos la lista vacía resultados. Luego iteramos sobre cada clave
    # en claves_rand. Accedemos al diccionario nombres_dicc con la clave actual
    # y añadimos el valor (que es la tupla) a la lista resultados usando
    # append(), y devolvemos la lista resultados.
    # Nota: los valores que toma clave en el bucle son los de la muestra
    # aleatoria que hemos generado.
    resultados = []
    for clave in claves_rand:
        resultados.append(nombres_dicc[clave])
    return resultados


# Llamamos a la función
num = 3
# Verificamos si el número introducido es mayor que el número de filas
if num > 245:
    print(f"Error: {num} es mayor que"
          " el número de nombres disponibles en el archivo (245).")
elif num < 1:
    print(f"Error: {num} es inválido.")
else:
    # Imprimimos el resultado
    resultado = random_asignation(df_nombres, num)
    print("Resultado:")
    print(resultado)
```

Para este ejercicio, primero se deberá crear una función para extraer la información del archivo .zip

Primero de todo se pide crear una función `zip_extraction` que recibirá el nombre del fichero y el nombre de la carpeta donde se quiere descomprimir. El zip a extraer se llama `Jorge_Luis.zip`. En este caso, el archivo .zip contiene ficheros de texto y que la búsqueda de la palabra se realiza sobre esos ficheros.

Para la segunda parte se pide crear una función `analyse_zip` que reciba el nombre del archivo zip y una tupla con las palabras que quiere que se busquen en todos los archivos que se encuentren en el zip.

Por lo que el resultado de la función es la suma de todas las veces que se encuentran las palabras que se indican en la tupla.

**Notas importantes:**

1) Para la segunda función se puede incluir la función `analyse_zip` dentro para poder descomprimir el archivo en caso de que lo necesite.

2) Utilizad los principios de **programación funcional** que hemos visto en teoría para resolver este ejercicio.

```python
# Respuesta
# Importamos zipfile, os, re y montamos el Drive.
import zipfile
import os
import re
from google.colab import drive
drive.mount('/content/drive')
archivo_zip = ('/content/drive/My Drive/Colab_Notebooks/'
               'Programación_DS/Activity_2/data/Jorge_Luis.zip')


def zip_extraction(zip_file, extract_to):
    """
    Extrae los archivos contenidos en un archivo ZIP a un directorio
    especificado. Recibe la ruta completa del archivo ZIP (zip_file) y la
    carpeta destino (extract_to). Si el archivo no existe o no es válido, se
    imprime un mensaje de error. No devuelve nada, pero extrae los
    archivos al directorio indicado.

    Usamos zipfile.ZipFile abrir el archivo ZIP en modo lectura ('r').
    Extraemos todos los archivos al directorio especificado con extractall.
    """
    with zipfile.ZipFile(zip_file, 'r') as zip_ref:
        zip_ref.extractall(extract_to)
    print(f"Archivos extraídos en: {extract_to}")


def find_text_files(directory):
    """
    La función devuelve una lista con las rutas de todos los archivos de
    texto (.txt) encontrados en un directorio y sus subdirectorios. Recibe
    como parámetro la ruta del directorio raíz desde donde se inicia
    la búsqueda.

    os.walk(directory) es una función que recorre todo un directorio y sus
    subdirectorios. Devuelve una tupla con tres valores:
    1) root: es el directorio actual que está siendo explorado.
    2) _: una lista de subdirectorios en el directorio actual
    (no la usamos en este caso, por eso ponemos un guion bajo).
    3) files: una lista de los nombres de los archivos dentro del
    directorio actual.

    os.path.join(root, file) toma el directorio (root) y el nombre del
    archivo (file) y los une en una ruta completa.
    Esto es necesario porque os.walk solo devuelve el nombre del archivo y no
    su ruta completa. Usamos os.path.join para asegurarnos de que la ruta sea
    correcta.
    """
    return [os.path.join(root, file)
            for root, _, files in os.walk(directory)
            for file in files if file.endswith(".txt")]


def analyse_zip(zip_file, search_words):
    """
    La función analiza los archivos de texto contenidos en un archivo ZIP,
    buscando coincidencias exactas de las palabras especificadas. Recibe la
    ruta del archivo ZIP (zip_file) y una tupla de palabras clave
    (search_words). Descomprime los archivos en un directorio recien creado, y
    cuenta las coincidencias exactas de las palabras en los archivos .txt.
    Devuelve el número total de coincidencias exactas encontradas.

    Nota: zip_dir es la ruta del directorio donde se almacenarán los archivos
    extraídos del zip. El valor de la variable es "extracted_zip", que es
    el nombre de la carpeta.
    """
    # Creamos la carpeta extracted_zip y extraemos los archivos usando la
    # función zip_extraction.
    zip_dir = "extracted_zip"
    zip_extraction(zip_file, zip_dir)

    # Encontramos todos los archivos de texto en la carpeta y subcarpetas
    # mediante la función find_text_files().
    text_files = find_text_files(zip_dir)

    # Inicializamos contador
    def count_word_reps(file_path):
        """
        La función cuenta las ocurrencias exactas de las palabras clave en un
        archivo .txt. Recibe la ruta del archivo (file_path) y devuelve el
        número de coincidencias exactas de las palabras en search_words,
        ignorando mayúsculas y minúsculas.

        Dentro de la función se usa open(file_path, 'r', encoding='utf-8')
        para abrir el archivo en modo lectura con soporte para caracteres
        especiales.
        file.read() lee todo el contenido del archivo.
        .lower() convierte el texto a minúsculas para ignorar diferencias
        entre mayúsculas y minúsculas.

        Se recorre cada palabra en search_words.
        Se usa re.findall con \b{word}\b para encontrar coincidencias exactas
        de la palabra completa, dónde \b es el delimitador de palabra.
        len(re.findall(...)) cuenta cuántas veces aparece esa palabra.
        La función acaba calculando el total de coincidencias de todas las
        palabras clave en el archivo y devuelve el total.
        """
        with open(file_path, 'r', encoding='utf-8') as file:
            content = file.read().lower()
            return sum(len(re.findall(rf'\b{word.lower()}\b', content))
                       for word in search_words)

    # Calculamos el total y lo devolvemos.
    total_count = sum(map(count_word_reps, text_files))
    return total_count


# Llamamos a la función y imprimimos el resultado
palabras_a_filtrar = ("la", "vida")
resultado = analyse_zip(archivo_zip, palabras_a_filtrar)
print(f"Total de coincidencias encontradas: {resultado}")
```
