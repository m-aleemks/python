### Ejercicio 2

Crea una matriz de 10x10 donde los valores de cada posición `(i,j)` correspondan a `abs(i*j-2)` para todo `i` múltiplo de `2`, `i+j` para todo `i` múltiplo de `5` y `(i+j)^2` para el resto de posiciones. **(1 punto)** $\color{green}{\text{NM}}$



```python
# Respuesta
# Importamos la libreria numpy y la nombramos como np mediante as
import numpy as np

# Creamos una matriz de 10x10
n = 10
matriz = np.zeros((n, n))

# Rellenamos la matriz según lo pedido: los valores de cada posición (i,j)
# correspondan a abs(i*j-2) para todo i múltiplo de 2, i+j para todo i múltiplo
# de 5 y (i+j)^2 (** para potencia, mientras que * sirve para multiplicar)
# para el resto de posiciones
for i in range(n):
    for j in range(n):
        if i % 2 == 0:
            matriz[i, j] = abs(i * j - 2)
        elif i % 5 == 0:
            matriz[i, j] = i + j
        else:
            matriz[i, j] = (i + j) ** 2

# Imprimimos la matriz
print(matriz)
```

### Ejercicio 3

Crea una función que genere una matriz de `0` y `1` simulando un tablero de ajedrez. La función debe cumplir con los siguientes requisitos:

- Debe preguntar al usuario cuál es la dimensión de la matriz. **(0.25 puntos)**
- La matriz que debe generar la función tiene que ser cuadrada.
- Debe mostrar el resultado visualmente en pantalla com si fuera una imagen en 2D. **(0.50 puntos)**

Verifica que la función se ejecuta correctamente probando con una matriz de `8x8`. **(0.25 puntos)**

**Nota**. Para visualizar la matriz tienes que buscar qué función de `Matplolib` te permite visualizar la matriz como una imagen en 2D.

**(Total 1 punto)** $\color{Orange}{\text{EG}}$

```python
# Respuesta
# Importamos la libreria numpy y matplotlib.pyplot
import numpy as np
import matplotlib.pyplot as mplt

# Definimos la función tablero_ajedrez y pedimos la dimensión de la matriz.
def tablero_ajedrez():

    dim = int(input("Introduce la dimensión del tablero: "))

    # Generamos una matriz cuadrada con ceros y tamaño dim x dim.
    # "dtype=int" especifica que los valores de la matriz serán enteros.
    # "tablero[1::2, ::2] = 1" asigna el valor 1 a las filas impares (1::2) y
    # las columnas pares (::2).
    # "tablero[::2, 1::2] = 1" asigna el valor 1 a las filas pares (::2) y
    # las columnas impares (1::2). La sintaxis de slicing es [start:stop:step].
    # Con tablero alternamos los valores entre 0 y 1, obteniendo el patrón de
    # un tablero de ajedrez.

    tablero = np.zeros((dim, dim), dtype=int)
    tablero[1::2, 0::2] = 1
    tablero[0::2, 1::2] = 1

    # "mplt.im.show()" muestra la imagen generada.
    mplt.imshow(tablero)

    # Este return tablero permite acceder a la matriz tablero fuera de la
    # función para usarla en otras partes del código.
    # Sin él, la matriz solo existiría dentro de la función.
    return tablero

# Probamos la función con un tablero 8x8 (fuera de la función)
tablero_8x8 = tablero_ajedrez()
```

### Ejercicio 4

Imagina que tienes tres cuerpos en un plano bidimensional.  Las fuerzas que actuan sobre cada cuerpo se pueden representar en una matriz.  Cada fila de la matriz representa un cuerpo y cada columna representa una fuerza en una dirección concreta ( per ejemplo, fuerza en la dirección X y fuerza en la dirección Y) en Newtons.

Las fuerzas que actuan sobre los cuerpos son las seguientes:
- Cuerpo A: Fuerza X = 10 y Fuerza Y = 5.
- Cuerpo B: Fuerza X = 4 y Fuerza Y = 12.
- Cuerpo C: Fuerza X = -6 y Fuerza Y = -8.

Responde a las seguientes preguntas:
- a) Crea la matriz de fuerzas y muéstrala por pantalla. **(0.25 puntos)**
- b) Calcula la fuerza resultante sobre cada cuerpo. La fuerza resultante se puede calcular como la suma vectorial de las fuerzas.  Para hacerlo, utiliza una función de `numpy`. **(0.50 puntos)**
- c) Calcula la aceleración de cada cuerpo mediante la segunda ley de Newton, `F=m*a`, considerando que la masa de cada cuerpo es de 5 kg. **(0.25 puntos)**

**Nota**.  Los resultados deben presentarse con sólo dos decimales.

**(Total 1 punto)** $\color{Orange}{\text{EG}}$

```python
# Respuesta
import numpy as np

# a) Representamos las fuerzas que actuan sobre los cuerpos
m_fuerza = np.array([[10, 5],
                    [4, 12],
                    [-6, -8]])
masa = 5

# Creamos la matriz de fuerzas y la mostramos
print("Matriz de fuerzas:")
print(m_fuerza)

# b) Calculamos la fuerza resultante sobre cada cuerpo usando la norma de
# cada fila. "np.linalg.norm" aplica la fórmula de la norma.
# Usamos axis=1 para indicar que el calculo se hace por filas.
fuerz_result = np.linalg.norm(m_fuerza, axis=1)
print("\nFuerzas resultantes (N):")
print(np.round(fuerz_result, 2))

# c) Calculamos la aceleración de cada cuerpo (a = F / m) y la imprimimos
aceleracion = fuerz_result / masa
print("\nAceleraciones (m/s^2):")
print(np.round(aceleracion, 2))
```

```python
# c)
# Extraemos el mes de la columna 'Date'
df['Month'] = df['Date'].dt.month

# Ventas totales por mes
ventas_mensuales = df.groupby('Month')['Total_Sales'].sum()

# ventas_mensuales.index representa los meses del eje X.
# ventas_mensuales.values representa las ventas totales del eje Y.
# marker='o' marca cada punto con un círculo.
# color='green' y linestyle='-' es el estilo y color de la línea.
plt.figure(figsize=(10, 6))
plt.plot(ventas_mensuales.index, ventas_mensuales.values, marker='o', color='green', linestyle='-')
plt.title("Ventas Totales a lo largo del tiempo")
plt.xlabel("Mes")
plt.ylabel("Ventas Totales")
plt.grid(True)
plt.show()
```

```python
# d)
# Calculamos la distribución de ventas por diferentes meses y visualizamos el
# resultado con un gráfico circular.
# plt.figure(figsize=(5, 5)) otorga un tamaño de un cuadrado 5x5.
# ventas_mensuales es un objeto que contiene las ventas totales por mes.
# kind='pie' especifica que queremos un gráfico circular.
# autopct='%1.1f%%' muestra los % resultantes con un decimal.
plt.figure(figsize=(5, 5))
ventas_mensuales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Distribución de Ventas por Mes")
plt.ylabel("")
plt.show()
```

### Ejercicio 6

En el siguiente ejercicio utilizaremos la librería `SciPy` para verificar si un conjunto de datos sigue una distribución normal.
   - a) Crea un conjunto de datos aleatorios con 50 valores que sigan una distribución normal ( media de 100 y desviación estándar de 15). **(0.50 puntos)** $\color{green}{\text{NM}}$
   - b) Realiza un test de normalidad utilizando la función [shapiro](https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.shapiro.html) para determinar si el conjunto de datos sigue una distribución normal. Muestra por pantalla el p-value y razona el resultado obtenido. **(1 punto)** $\color{orange}{\text{EG}}$
   - c) Crea un histograma del conjunto de datos para visualizar su distribución. **(0.50 puntos)** $\color{green}{\text{NM}}$  
   - d) Utiliza una función de la librería `seaborn` per visualizar la función de densidad de probabilidad correspondiente a la distribución de los datos creados. **Opcional**. $\color{red}{\text{EI}}$  
   


**(Total 2 puntos)**

```python
# Respuesta
# a)
import numpy as np

# Creamos el conjunto de datos pedido usando random y normal.
mean = 100
sd = 15
datos = np.random.normal(loc=mean, scale=sd, size=50)

print("Conjunto de datos:")
print(datos)
```

```python
# b)
# Importamos la función shapiro y realizamos el test de Shapiro-Wilk
from scipy.stats import shapiro

# shapiro(datos) realiza el test de normalidad sobre el array o lista datos.
# stat es el valor del estadístico W del test. Este valor mide la desviación de
# los datos de una distribución normal. Cuanto más cerca de 1 esté el valor,
# más probable es que los datos sigan una distribución normal.
# p_value es el valor p asociado con el test, que dependiendo de su valor nos
# indicará si los datos pueden ser considerados normalmente distribuidos.
stat, p_value = shapiro(datos)
print("\nResultados del test de Shapiro-Wilk:")
print(f"Estadístico W: {stat:.4f}, p-value: {p_value:.4f}")

# Interpretamos los resultados
alpha = 0.05
if p_value > alpha:
    print("No se puede rechazar la Ho. Los datos siguen una distribución normal.")
else:
    print("Se rechaza la Ho. Los datos no siguen una distribución normal.")
```

```python
# c)
# Creamos el histograma con los datos aleatorios. Son normales si siguen la
# forma de una campana, característico de una distribución normal.
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(datos, edgecolor='black')
plt.title("Histograma de los datos")
plt.xlabel("Valor")
plt.ylabel("Frecuencia")
plt.grid(axis='y', alpha=0.75)
plt.show()
```
