## Ejercicio 1
Estudiaremos la complejidad algorítmica en relación al tiempo de ejecución de la siguiente función, tendréis que calcular la complejidad algorítmica Θ y justificar vuestra respuesta. Podéis asumir que todos los parámetros de entrada tienen el mismo tamaño.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1 punto)**

```python
def algoritmicComplexity(L: list[int], m: list[int], d: dict[str, int]):
    """
    Args:
        This is a method for calculating the algoritmic complexity and it
        received the following arguments
        L: a list of integers
        m: another list of integers
        d: a dictionary with the same length than L and s
    """
    r = 0
    for k in d.keys():
        for i, j in zip(L, m):
            d[k] = i * j
            r = d[k] + 1
    return r
```

Respuesta:

```python
# Respuesta
# Importamos timeit y numpy como np. timeit sirve para medir
# el tiempo de ejecución de funciones o fragmentos de código.
# numpy maneja datos numéricos y generar listas de números aleatorios.
import timeit
import numpy as np


# Análisis de la complejidad de la función proporcionada
# La función tiene dos bucles anidados:
# El primer bucle recorre las claves del diccionario d, lo que implica O(n),
# donde n es el número de claves en d.
# El segundo bucle recorre las listas L y m de tamaño m, implicando O(m).
# Así, la complejidad total del algoritmo es:
# Theta(n * m), donde:
# n es el número de claves en el diccionario d.
# m es una lista cuyos elementos se emparejan con los de L.
# El tamaño de las listas L y m afecta el número de iteraciones del
# bucle interno.
def algoritmicComplexity(L: list[int], m: list[int], d: dict[str, int]):
    """
    Args:
        This is a method for calculating the algoritmic complexity and it
        received the following arguments
        L: a list of integers
        m: another list of integers
        d: a dictionary with the same length than L and s
    """
    r = 0
    for k in d.keys():  # Itera sobre las claves del diccionario
        # zip(L, m) para empareja elementos de las listas L y m
        # En O(m), m es el tamaño de las listas.
        for i, j in zip(L, m):  # Itera sobre pares (i, j) de las listas L y m
            d[k] = i * j  # Calcula el producto de i y j y lo asigna a d[k]
            r = d[k] + 1  # Actualiza el valor de r con el producto + 1
    return r


# Generamos datos de prueba
sizes = [10, 50, 100, 200]  # Diferentes tamaños de entrada para análisis
for size in sizes:
    L = list(np.random.randint(1, 100, size=size))  # Lista L de tamaño 'size'
    m = list(np.random.randint(1, 100, size=size))  # Lista m de tamaño 'size'
    # Diccionario con 'size' claves inicializadas a 0
    d = {f"key_{i}": 0 for i in range(size)}

    # Medimos el tiempo de ejecución con timeit
    # Creamos el Timer que ejecutará la función con los datos generados
    t = timeit.Timer(lambda: algoritmicComplexity(L, m, d))

    # Ejecutamos la función 10 veces, repetimos 5 veces, y escogemos el mín.
    tiempo_min = min(t.repeat(number=10, repeat=5))

    # Imprimimos el tamaño de la entrada y el tiempo mínimo
    print(f"Tamaño de entrada {size}:"
          f" tiempo para 10 ejecuciones es {tiempo_min:.5f} segundos")
```

51.50853, -0.12574 lo cual coincide con los primeros 12 dígitos que has mostrado por pantalla. ¿Podrías indicarles cuales de las siguientes ciudades aparecen en el fichero? ¿Podrías indicar también la posición del fichero en la que aparecen?

-Barcelona: 41.3851° N, 2.1734 (41385121734)

**Solución 2.2**

```python
# Respuesta
# Definimos las coordenadas como cadenas para facilitar la búsqueda
cities = {
    "Barcelona": "41385121734",
    "Madrid": "40416837038",
    "Sevilla": "37388659823"
}

# Leemos el contenido del fichero
with open("randomNumbers.txt", "r") as file:
    data = file.read()

# Buscamos las coordenadas en el fichero
# coordinates es la cadena asociada a cada ciudad en el diccionario
# find() se usa para buscar una subcadena en una cadena. Si encuentra la
# subcadena, devuelve el índice (posición) donde empieza esa subcadena
# en la cadena principal. Si no encuentra la subcadena, devuelve -1.
for city, coordinates in cities.items():
    position = data.find(coordinates)  # Buscamos la posición de la secuencia
    if position != -1:
        print(f"{city} aparece en el fichero, empezando"
              f"en la posición {position}.")
    else:
        print(f"{city} no aparece en el fichero.")
```

### 2.3    (0.5 puntos)
Seguramente, habrás notado que has tenido que hacer el mismo código tres veces, una para Madrid, otra para Barcelona y otro para Sevilla. Dividir el código en módulos o funciones nos ayuda a reutilizar el código y a ahorrar trabajo a los programadores. ¿Crees que podrías crear una función que pueda reutilizar el código?

A esta función le pasaremos dos parámetros de entrada, uno sería la ruta del fichero y el otro serían las coordenadas que quieres buscar. Esta función no devolverá ningún parámetro pero sí que mostrará por pantalla si ese patrón aparece en el fichero y en qué posición está.

De esta forma, si en el futuro quieren buscar nuevas ciudades siempre se podría reutilizar el mismo código y ser más eficiente.

```python
# Respuesta
def search_coordinates_in_file(file_path: str, coordinates: str):
    """
    Busca una secuencia de coordenadas en un fichero y muestra si aparece y en
    qué posición. Args: file_path (str) es la ruta del fichero donde buscar.
    coordinates (str) es la secuencia de coordenadas (como cadena de dígito)
    a buscar. Returns: None, imprime los resultados en pantalla.
    """
    # Leemos el contenido del fichero
    with open(file_path, "r") as file:
        data = file.read()

    # Buscamos las coordenadas en el contenido
    position = data.find(coordinates)

    # Imprimimos resultados
    if position != -1:
        print(f"Las coordenadas {coordinates} aparecen en el fichero,"
              f"empezando en la posición {position}.")
    else:
        print(f"Las coordenadas {coordinates} no aparecen en el fichero.")


# Ruta del fichero
file_path = "randomNumbers.txt"

# Coordenadas de las ciudades
cities = {
    "Barcelona": "41385121734",
    "Madrid": "40416837038",
    "Sevilla": "37388659823"
}

# Usamos la función para buscar cada ciudad
for city, coordinates in cities.items():
    print(f"Buscando en el fichero para {city}:")
    search_coordinates_in_file(file_path, coordinates)
    print()
```

Los espías han conseguido descifrar que el fichero contiene coordenadas con una lista de ciudades que el enemigo tiene en su punto de mira, sin embargo, se han encontrado un obstáculo: la programación no es su punto fuerte. En este ejercicio detectaréis las ineficiencias y mejoraréis el código.

1. Utilizando las herramientas de *profiling* que hemos visto en el Notebook de teoría, analizad el siguiente fragmento de código ineficiente y encontrad cual es el cuello de botella. Explicad las ineficiencias que habéis detectado, relacionándolas con los resultados obtenidos en el *profiling*.

2. Mejorad el código con el objetivo de reducir el tiempo de ejecución.

<span style="font-family: Courier New; background-color: #82b74b; color: #000000; padding: 3px; ">SM</span> **(1.5 puntos)**

```python
from random import randint


def count_appearance(iters, n):
    """
     Given a number n, generates a random list of 'iters'
     number of elements between [-10*n,10*n] and count how many times
     our input n appears in the random list.

     Args:
         iters (:obj: `int`) : integer, number of random elements to generate
         n (:obj: `int`) : integer, number to count appearences
     Returns:
         (:obj: `int`) : integer, number of times that n
         appears in the generated random list.
    """
    c = 0
    L = []
    for i in range(iters):
        L.append(randint(-10*n, 10*n))
    for j in L:
        if j == n:
            c += 1
    return c


iters, n = int(1e6), int(1e2)
count_appearance(iters, n)
```

Respuesta:

```python
# Respuesta
# Utilizando las herramientas de profiling que hemos visto en la
# de teoría, analizamos el código original y detectamos el cuello de
# botella. Observamos que el tiempo de ejecución se veía afectado
# principalmente por la generación y almacenamiento de una lista de
# gran tamaño (L), que no era estrictamente necesaria. Este almacenamiento
# incrementaba significativamente el uso de memoria y tiempo, ya que cada
# elemento de la lista era evaluado de nuevo en un bucle separado.

# El profiling muestra que la función "append" para añadir elementos a la
# lista y el segundo bucle que recorría "L" eran los componentes más
# costosos en términos de tiempo. Como solución, eliminamos la necesidad
# de almacenar la lista completa al usar una expresión que
# cuenta directamente las apariciones del número objetivo durante la
# generación de números aleatorios, reduciendo tanto el uso de memoria
# como el tiempo de ejecución.
from random import randint
from timeit import Timer


def optimized_count_appearance(iters, n):
    """
    Genera iters números aleatorios entre -10*n y 10*n y cuenta
    cuántas veces el número n aparece entre ellos eficientemente
    sin almacenar todos los números generados, sino simplemente
    sumando 1 cada vez que el número generado es igual a n.
    Devuelve la cantidad total de veces que n ha aparecido
    en los iters intentos.
    """
    # El "_" se usa como un marcador de lugar para indicar que el valor
    # generado por range(iters) no se va a utilizar, i.e., estamos iterando
    # iters veces, pero el valor actual de la iteración no tiene importancia,
    # por lo que se le asigna el nombre _ en lugar de uno más descriptivo.
    return sum(1 for _ in range(iters) if randint(-2 * n, 2 * n) == n)


# Parámetros
iters = int(1e6)
n = int(1e2)

# Medimos tiempos usando timeit.Timer
# Configuramos el Timer para cada función
t_original = Timer(lambda: count_appearance(iters, n))
t_optimizado = Timer(lambda: optimized_count_appearance(iters, n))

# Ejecutamos la instrucción 10 veces y repetimos 5 veces
tiempos_original = t_original.repeat(number=10, repeat=5)
tiempos_optimizado = t_optimizado.repeat(number=10, repeat=5)

# Seleccionamos el valor mínimo como recomendación de la documentación
mejor_tiempo_original = min(tiempos_original)
mejor_tiempo_optimizado = min(tiempos_optimizado)

# Ejecución de ambas funciones para obtener resultados
resultado_original = count_appearance(iters, n)
resultado_optimizado = optimized_count_appearance(iters, n)

# Imprimimos resultados
print(f"Función Original -> Resultado: {resultado_original},"
      f" Mejor Tiempo: {mejor_tiempo_original:.5f} segundos")
print(f"Función Optimizada -> Resultado: {resultado_optimizado},"
      f" Mejor Tiempo: {mejor_tiempo_optimizado:.5f} segundos")
```

<span style="font-family: Courier New; background-color: #ffcc5c; color: #000000; padding: 3px; ">EG</span> **(1 punto)**

Por último, también necesitan ayuda con el siguiente código, tu objetivo será detectar las eficiencias y mejorarlo.

Para eso podéis utilizar las herramientas de profiling que tenéis en el Notebook de teoría. Debéis analizar las ineficiencias e identificar los cuellos de botella. Explicar también las indeficiencias detectadas en base a los resultados obtenidas en el profiling.

```python
def squaresum(n: int):
    """devuelve la suma de los cuadrados para los primeros n números naturales.

    Parámetros de entrada:
        n: número entero

    Valores que devuelve:
        (int): suma de cuadrados de 1 a n.
    """
    total = 0
    for i in range(1, n + 1):  # Itera de 1 a n (range excluye lím. sup.).
        # Itera i veces.
        for j in range(i):  # Bucle innecesario para calcular el cuadrado.

            total += i
    return total

# En lugar de calcular directamente i^2, la función suma repetidamente el
# número i, lo que hace que el proceso sea más lento. Complejidad: O(n^2).
```

**Respuesta:**

```python
# Respuesta
import timeit


def squaresum_optim(n: int):
    """
    Devuelve la suma de los cuadrados para los primeros n números naturales
    usando una fórmula directa. Parámetros de entrada n (número entero).
    Devuelve (int) suma de cuadrados de 1 a n.
    """
    return (n*(n+1) * (2*n+1)) // 6
# Usamos una fórmula matemática directa para calcular la suma
# de los cuadrados de los primeros n números naturales
# (sin bucles, y de complejidad O(1)).


# Configuración de "n"
n = 10_000

# Configuramos los timers para ambas funciones
t_original = timeit.Timer(lambda: squaresum(n))
t_optimizado = timeit.Timer(lambda: squaresum_optim(n))

# Ejecutamos las funciones 10 veces y repetimos 5 veces
tiempos_original = t_original.repeat(number=10, repeat=5)
tiempos_optimizado = t_optimizado.repeat(number=10, repeat=5)

# Seleccionamos el valor mínimo como el más representativo
mejor_tiempo_original = min(tiempos_original)
mejor_tiempo_optimizado = min(tiempos_optimizado)

# Imprimimos los resultados
print(f"Tiempo código original: {mejor_tiempo_original:.6f} segundos")
print(f"Tiempo código optimizado: {mejor_tiempo_optimizado:.6f} segundos")
```

Multithreading ya que:

- Consultar la API (externa) de Open Library es una tarea I/O-bound (operación de entrada/salida, dependiente de la latencia y velocidad de la red). Durante el tiempo que se espera una respuesta, el thread puede manejar otras solicitudes sin desperdiciar tiempo de CPU.
- Multiprocessing no aporta beneficios significativos aquí porque el cuello de botella no está en el cálculo intensivo.
- Los threads (subprocesos) son más adecuados para tareas I/O-bounded, ya que permiten la ejecución concurrente mientras se espera una respuesta de la red.
- Crear procesos en multiprocessing tiene un costo mayor en comparación con threads, especialmente para tareas que no requieren cálculos intensivos de CPU (e.g., las tareas I/O-bound no utilizan completamente la CPU).
- Multithreading permite manejar múltiples solicitudes de API de manera concurrente con menos recursos y complejidad. Si el servicio planea realizar consultas frecuentes, threads es más eficiente en este caso.

```python
# Respuesta
import requests
from concurrent.futures import ThreadPoolExecutor
import timeit


def get_book_info(isbn):
    """Devuelve título, autores y URL de la portada a partir de un ISBN
    usando la API de Open Library."""
    openLibraryUrl = "https://openlibrary.org/api/books"
    parameters = {
        "bibkeys": f"ISBN:{isbn}",
        "format": "json",
        "jscmd": "data",
    }
    response = requests.get(openLibraryUrl, params=parameters)

    # Comprobamos si la respuesta contiene datos para el ISBN
    if response.status_code != 200 or f"ISBN:{isbn}" not in response.json():
        # Si no se encuentra el libro, devolvemos valores predeterminados
        return (f"Información no disponible para ISBN: {isbn}", [], "")

    book_info = response.json()[f"ISBN:{isbn}"]
    title = book_info["title"]
    authorNames = []
    for author in book_info["authors"]:
        authorNames.append(author["name"])
    coverImageUrl = (
        book_info["cover"]["small"] if "cover" in book_info else ""
    )

    return (title, authorNames, coverImageUrl)


def get_books_with_threadpoolexecutor(isbns, max_threads=11):
    """
    Consulta los datos de libros de forma concurrente con ThreadPoolExecutor.
    """
    start_time = time.time()
    with ThreadPoolExecutor(max_threads) as executor:
        results = list(executor.map(get_book_info, isbns))
    elapsed_time = time.time() - start_time
    return elapsed_time, results


# Lista de ISBNs para prueba
isbns_to_test = [
    "0134692888", "1491946008", "1491957662", "0134853989",
    "1449340377", "1449355730", "1491939362", "1775093301",
    "1593279280", "1593279922"
]

# Comparamos tiempos con diferentes números de threads (1 a 10)
for num_threads in range(1, 11):
    # Configuramos el Timer para medir la ejecución de la función
    t = timeit.Timer(lambda: get_books_with_threadpoolexecutor
                     (isbns_to_test, max_threads=num_threads))

    # Ejecutamos la función 10 veces, repetimos 5 veces, y escogemos el mín.
    mejor_tiempo = min(t.repeat(number=10, repeat=5))

    print(f"Tiempo con {num_threads} threads: {mejor_tiempo:.4f} segundos")

# Llamamos a la función y obtenemos los resultados
elapsed_time, resultados = (get_books_with_threadpoolexecutor
                            (isbns_to_test, max_threads=8))
```
